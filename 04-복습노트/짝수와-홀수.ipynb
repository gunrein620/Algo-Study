## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(num):
    return "Even" if num % 2 == 0 else "Odd"

print(solution_model(0))   # Even
print(solution_model(-3))  # Odd
print(solution_model(4))   # Even
```
