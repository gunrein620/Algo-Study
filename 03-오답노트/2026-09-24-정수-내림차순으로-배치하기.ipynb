## 해설

### 풀이 메모

내 실수: map(int, ...)로 정수들을 만든 뒤 join에 넣었다.

join은 문자열끼리 합친다. 모두 합친 뒤 마지막에 int로 바꾼다.

sorted는 리스트 반환, join은 문자열 반환, int는 정수 반환.


## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(n):
    return int(''.join(sorted(str(n), reverse=True)))

print(solution_model(118372))  # 예상 출력: 873211
```
