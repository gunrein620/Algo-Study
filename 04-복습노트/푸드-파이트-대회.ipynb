## 해설

### 먼저 떠올리기

- 한 선수에게 주는 음식 개수는 왜 `count // 2`일까?
- 정수 리스트를 `join()`하려면 무엇을 바꿔야 할까?
- 왼쪽 배치만 만들면 오른쪽은 어떻게 구할까?

`food[1:]`로 물을 제외하고, `start=1`로 음식 번호를 1부터 붙인다.
`[1, 3, 4, 6]`이면 `'1' * 1 + '2' * 2 + '3' * 3`으로 왼쪽은 `'122333'`.
오른쪽은 `left[::-1]`인 `'333221'`. 가운데 `'0'`을 끼운다.

접근은 맞았다. 처음에는 정수 리스트를 그대로 `join()`해서 타입 오류가 났고, `map(str, answer)`로 수정했다.
`left`가 음식 번호 오름차순이라 역정렬도 맞지만, 대칭 배치라는 의도는 `left[::-1]`이 더 직접적이다.

| 내 풀이 | 모범답안 |
| --- | --- |
| while로 번호를 하나씩 추가 | `str(idx) * 개수`로 반복 |
| 역정렬로 오른쪽 생성 | `[::-1]`로 왼쪽 반전 |
| 마지막에 정수 → 문자열 변환 | 처음부터 문자열로 생성 |

핵심: **절반씩 나누기 → 왼쪽 만들기 → 왼쪽 + 물 + 뒤집은 왼쪽**.


## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(food):
    left = ''
    for idx, count in enumerate(food[1:], start=1):
        left += str(idx) * (count // 2)
    return left + '0' + left[::-1]

print(solution_model([1, 3, 4, 6]))  # 1223330333221
print(solution_model([1, 7, 1, 2]))  # 111303111
```

### 기존 풀이 비교 1

기존 노트에 기록된 다른 풀이입니다.


```python
def solution_compare_1(food):
    answer = []
    left = []

    for idx, i in enumerate(food):
        if idx == 0:
            continue
        amount = i // 2
        while amount > 0:
            left.append(idx)
            amount -= 1

    right = sorted(left, reverse=True)
    left.append(0)
    answer = left + right
    return ''.join(map(str, answer))

print(solution_compare_1([1, 3, 4, 6]))  # 1223330333221
print(solution_compare_1([1, 7, 1, 2]))  # 111303111
```
