# zip과 reversed — 함께 읽기와 거꾸로 읽기

[전체 목차](../../README.md) · [도구 목차](../README.md) · [이 장 목차](README.md)

**이 장의 목표:** 두 배열의 같은 위치는 zip으로, 한 배열의 역순은 reversed로 처리한다.
1~4절은 zip, 5~6절은 reversed입니다. 각 코드 셀은 독립 실행할 수 있습니다.

## 1. zip 개념과 모범 코드 — 같은 위치끼리 짝짓기

`zip(a, b)`는 두 입력에서 같은 위치에 있는 원소를 한 쌍으로 제공합니다.
`for x, y`는 그 쌍을 x와 y에 나누어 받습니다. x와 y는 **값**입니다.

**이건 코테에서 자주 쓰니 챙겨라: 같은 위치의 두 값을 함께 쓰면 `zip`이다.**


```python
def dot_product_model(a, b):
    return sum(x * y for x, y in zip(a, b))

print(dot_product_model([1, 2, 3], [4, 5, 6]))  # 예상 출력: 32
print(list(zip([1, 2, 3], [4, 5, 6])))  # 예상 출력: [(1, 4), (2, 5), (3, 6)]
```

모범 코드의 조건은 두 배열의 길이가 같다는 것입니다.
`x * y for ...`는 곱셈 결과를 하나씩 제공하는 제너레이터 표현식이고, 바깥의 sum이 그 결과를 더합니다.

| 반복 | x | y | x * y | 누적 합 |
| --- | --- | --- | --- | --- |
| 1 | 1 | 4 | 4 | 4 |
| 2 | 2 | 5 | 10 | 14 |
| 3 | 3 | 6 | 18 | 32 |

일반 for문으로 풀면 두 단계가 선명해집니다.


```python
a = [1, 2, 3]
b = [4, 5, 6]
total = 0
for x, y in zip(a, b):
    product = x * y
    total += product
print(total)  # 예상 출력: 32
```

`a`, `b`는 입력 배열이고 `x`, `y`는 이번 반복에서 꺼낸 원소입니다.
이 예제에서 x=1을 `a[x]`에 넣으면 첫 번째 값이 아니라 a의 인덱스 1에 있는 2를 읽게 됩니다.

## 2. 중첩 for문과 zip은 짝짓는 방식이 다르다

zip은 같은 위치의 짝만 만듭니다. 중첩 for문은 왼쪽 원소 하나마다 오른쪽을 전부 순회합니다.


```python
a = [1, 2]
b = [10, 20]
paired = list(zip(a, b))
all_pairs = []
for x in a:
    for y in b:
        all_pairs.append((x, y))
print(paired)  # 예상 출력: [(1, 10), (2, 20)]
print(all_pairs)  # 예상 출력: [(1, 10), (1, 20), (2, 10), (2, 20)]
```

`(1, 20)`까지 필요한 문제라면 중첩 반복, 위치가 대응하는 문제라면 zip입니다.
원소 수가 각각 n, m이면 모든 쌍은 n×m개이고 zip의 쌍은 min(n, m)개입니다.

## 3. 모범 코드 — 절댓값에 같은 위치의 부호 붙이기

두 배열의 길이가 같고 signs의 원소는 bool이라고 가정합니다.
True이면 양수, False이면 음수로 해석해 합산합니다. 이 경우는 분기가 보이는 반복문이 읽기 좋습니다.


```python
def signed_sum_model(absolutes, signs):
    total = 0
    for number, positive in zip(absolutes, signs):
        if positive:
            total += number
        else:
            total -= number
    return total

print(signed_sum_model([4, 7, 12], [True, False, True]))  # 예상 출력: 9
print(signed_sum_model([], []))  # 예상 출력: 0
```

| number | positive | 처리 | total |
| --- | --- | --- | --- |
| 4 | True | 4 더하기 | 4 |
| 7 | False | 7 빼기 | -3 |
| 12 | True | 12 더하기 | 9 |

이미 합을 구하고 있으므로 결과 리스트에 append한 뒤 다시 합산할 필요는 없습니다.
익숙해지면 같은 계산을 다음처럼 읽을 수 있습니다.


```python
absolutes = [4, 7, 12]
signs = [True, False, True]
answer = sum(number if positive else -number
             for number, positive in zip(absolutes, signs))
print(answer)  # 예상 출력: 9
```

`number if positive else -number`는 True일 때 number, False일 때 -number를 선택합니다.
지금은 위의 일반 for문을 정확하게 읽을 수 있으면 충분합니다.

## 4. zip 문법·반환·경계 조건

`zip(a, b, ...)`는 순회 가능한 입력 여러 개를 받아, 원소를 튜플로 묶어 제공하는 **zip 객체(iterator)**를 반환합니다.
입력 리스트는 바꾸지 않습니다. 입력이 iterator이면 그 진행 위치는 바뀝니다.
기본 zip은 **가장 짧은 입력이 끝나면 종료**합니다. 남는 원소를 자동으로 채우지 않습니다.


```python
print(list(zip([1, 2], ['a'])))  # 예상 출력: [(1, 'a')]
print(list(zip([], [1])))  # 예상 출력: []
print(list(zip()))  # 예상 출력: []

pairs = zip([1, 2], ['a', 'b'])
print(type(pairs).__name__)  # 예상 출력: zip
print(list(pairs))  # 예상 출력: [(1, 'a'), (2, 'b')]
print(list(pairs))  # 예상 출력: []
```

길이가 같아야 정답인 문제는 문제의 길이 보장을 확인합니다. 직접 입력을 검증해야 하는 상황이라면 리스트의 길이를 비교하세요.
`zip(3, [1])`은 정수 3을 순회할 수 없어 `TypeError`입니다.

| 필요한 짝 | 도구 |
| --- | --- |
| 번호 + 한 입력의 값 | `enumerate(values)` |
| 첫 배열의 값 + 두 번째 배열의 값 | `zip(a, b)` |
| 번호 + 두 배열의 값 | `enumerate(zip(a, b))` — 나중에 조합해도 됨 |

## 5. reversed 개념과 모범 코드 — 뒤에서부터 읽기

`reversed(values)`는 뒤에서부터 원소를 제공합니다. **정렬하지 않고 순서만 반대로 읽습니다.**
리스트·문자열·튜플·range처럼 역순으로 읽을 수 있는 대상을 받습니다.

**이건 코테에서 자주 쓰니 챙겨라: 역순 순회는 `reversed`, 내림차순 정렬은 `sorted(..., reverse=True)`다.**


```python
values = [3, 1, 4]
backward = list(reversed(values))
descending = sorted(values, reverse=True)
print(backward)  # 예상 출력: [4, 1, 3]
print(descending)  # 예상 출력: [4, 3, 1]
print(values)  # 예상 출력: [3, 1, 4]
print(''.join(reversed('abc')))  # 예상 출력: cba
```

| 원본 위치 | 역순으로 읽을 때 순서 | 값 |
| --- | --- | --- |
| 2 | 첫 번째 | 4 |
| 1 | 두 번째 | 1 |
| 0 | 세 번째 | 3 |

reversed도 리스트 자체가 아니라 iterator를 반환합니다. 리스트가 필요하면 list, 문자열이면 join으로 모읍니다.
원본은 변경하지 않으며 원소의 타입도 유지합니다. 일반 반복문으로 직접 읽어도 됩니다.


```python
values = [3, 1, 4]
answer = []
for value in reversed(values):
    answer.append(value)
print(answer)  # 예상 출력: [4, 1, 3]
print(values)  # 예상 출력: [3, 1, 4]
```

## 6. reversed·reverse·슬라이싱 비교

| 표현 | 반환/결과 | 원본 리스트 변경 |
| --- | --- | --- |
| `reversed(arr)` | 역순으로 읽는 iterator | 없음 |
| `list(reversed(arr))` | 역순의 새 list | 없음 |
| `arr[::-1]` | 역순의 새 list(문자열이면 새 str) | 없음 |
| `arr.reverse()` | 반환값 None | 있음 |


```python
values = [3, 1, 4]
returned = values.reverse()
print(values)  # 예상 출력: [4, 1, 3]
print(returned)  # 예상 출력: None
print(list(reversed([])))  # 예상 출력: []
print(''.join(reversed('')) == '')  # 예상 출력: True

backward = reversed([1, 2])
print(list(backward))  # 예상 출력: [2, 1]
print(list(backward))  # 예상 출력: []
```

`answer = arr.reverse()`를 쓰면 answer는 None입니다. 역순 결과를 따로 저장하려면 `arr[::-1]` 또는 `list(reversed(arr))`를 씁니다.

reversed는 아무 iterable이나 받지는 않습니다. 예를 들어 일반 제너레이터나 map 객체는 뒤에서 읽는 기능이 없어 `TypeError`입니다.
필요하다면 `saved = list(대상)`으로 먼저 저장한 뒤 `reversed(saved)`를 사용할 수 있습니다.
이때는 저장 공간과 원래 iterator의 소비가 발생합니다.

## 7. 가져갈 것

1. 같은 위치의 두 값은 zip으로 짝짓는다. 중첩 for문은 모든 조합이다.
2. zip에서 나온 x, y는 원소 값이다. 인덱스로 다시 사용하지 않는다.
3. 역순 읽기, 내림차순 정렬, 원본 뒤집기를 구분한다.

입력 길이가 n, m이고 원소 처리가 O(1)일 때 zip 전체 순회는 O(min(n, m)), zip 자체 공간은 O(1)입니다(입력 개수 고정).
위 내적·부호 합산의 추가 공간도 O(1)입니다. 일반 리스트 n개를 reversed로 모두 순회하면 시간 O(n), iterator 공간 O(1)입니다.
`list(reversed(arr))`와 리스트 슬라이싱은 새 결과 공간 O(n)을 사용합니다.
