## 해설

기록일: 2026-09-28 · 카드 뭉치 문제에서 배운 표현. 설명 후 이해를 표현했으며 독립 재풀이는 미확인.

### 먼저 떠올리기

1. `if cards1 and word == cards1[0]:`에서 두 조건은 각각 무엇인가?
2. `cards1 = []`일 때 왜 `[0]` 접근 오류가 나지 않을까?
3. 조건 순서를 바꿔도 안전할까? `[0]`은 빈 리스트일까?

```python
cards1 = ["i", "drink"]
word = "i"

if cards1 and word == cards1[0]:
    print("사용 가능")  # 사용 가능
```

**“카드가 남아 있고, 맨 앞 카드가 원하는 단어라면”**이라는 뜻이다. 리스트와 단어를 함께 비교하는 것이 아니라 두 조건을 연결한다.

리스트에 대해서는 아래처럼 풀어 읽을 수 있다.

```python
cards1 = ["i", "drink"]
word = "i"

if len(cards1) > 0:
    if word == cards1[0]:
        print("사용 가능")  # 사용 가능
```

빈 리스트는 조건식에서 거짓으로 판단한다. **and는 왼쪽이 거짓이면 오른쪽을 실행하지 않는다.** 이를 단락 평가라고 하며, 코테에서 자주 쓰니 챙긴다.

| `cards1` (list) | `word` (str) | 왼쪽 판단 | 오른쪽 비교 | if 본문 실행 |
| --- | --- | --- | --- | --- |
| `[]` | `"i"` | 거짓 | 실행하지 않음 | 안 함 |
| `["i", "drink"]` | `"i"` | 참 | `"i" == "i"` → 참 | 실행 |
| `["i", "drink"]` | `"want"` | 참 | `"want" == "i"` → 거짓 | 안 함 |

```python
cards1 = []
word = "i"

if cards1 and word == cards1[0]:
    print("사용 가능")
else:
    print("사용 불가")  # 사용 불가
```

`word == cards1[0] and cards1`로 순서를 바꾸면 빈 리스트의 `[0]`부터 읽어서 `IndexError`가 난다. **존재 여부를 먼저 확인한 뒤 원소에 접근한다.** 조건식 자체는 원본 리스트를 변경하지 않는다.

`[0]`은 원소가 한 개 있으므로 참이다. 원소의 값이 아니라 리스트가 비었는지를 판단한다.

주의: 리스트 자체가 `False`로 바뀌는 것은 아니다. `and`는 항상 bool을 반환하지 않고 피연산자 값을 반환한다.

```python
cards1 = []
word = "i"
result = cards1 and word == cards1[0]
print(result)        # []
print(type(result))  # <class 'list'>
```

이 결과 `[]`를 `if`가 거짓으로 판단하는 것이다. 기존에 배운 `or`는 반대로 왼쪽이 참이면 오른쪽을 실행하지 않는다. 다른 빈 값과 기본값 선택은 [도구 설명](../01-파이썬-도구/06-문법/truthiness.md)에서 복습한다.

### 전체 풀이로 연결

`goal`을 한 단어씩 읽으면서 두 뭉치의 맨 앞 카드와 비교한다. 두 뭉치 모두 해당하지 않으면 `No`, 모든 단어를 만든 뒤에만 `Yes`를 반환한다. 아래 풀이는 복사본에서 `pop(0)`을 사용해 기존에 배운 `빈 리스트와 and`를 그대로 적용한다. 원본 배열은 바꾸지 않는다.

`a`, `b`를 두 카드 뭉치의 길이, `g`를 목표 단어 수라 하면, 리스트 맨 앞 삭제는 뒤 원소를 이동시키므로 최악 시간은 O(a + b + g × max(a, b)), 복사본 공간은 O(a + b)다. 이 문제에서는 각 뭉치 길이가 최대 10이다. 문자열 비교 비용은 단어 길이 최대 10인 조건에서 상수로 본다.


## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


````python
def solution_model(cards1, cards2, goal):
    # 원본 카드 배열을 바꾸지 않도록 복사한다.
    cards1 = cards1.copy()
    cards2 = cards2.copy()

    for word in goal:
        if cards1 and word == cards1[0]:
            cards1.pop(0)
        elif cards2 and word == cards2[0]:
            cards2.pop(0)
        else:
            return "No"

    return "Yes"

print(solution_model(['i', 'drink', 'water'], ['want', 'to'], ['i', 'want', 'to', 'drink', 'water']))  # 예상 출력: Yes
print(solution_model(['i', 'water', 'drink'], ['want', 'to'], ['i', 'want', 'to', 'drink', 'water']))  # 예상 출력: No
````
