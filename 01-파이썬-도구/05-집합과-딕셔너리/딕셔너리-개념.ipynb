## 1. 딕셔너리는 이름표로 값을 찾는다

### 개념

리스트는 `arr[0]`처럼 위치로 찾습니다. 딕셔너리는 `prices['사과']`처럼 **키**로 찾습니다.

`{'사과': 1000, '배': 2000}`에서 콜론 왼쪽은 키, 오른쪽은 값입니다. 여기서는 이름(str)이 키이고 가격(int)이 값입니다. `{}`는 빈 딕셔너리입니다.

`d[키]`는 저장된 값을 반환하며 원본을 바꾸지 않습니다. 반환 타입은 저장된 값의 타입입니다. **이건 코테에서 자주 쓰니 챙겨라: `d[key]`는 키 → 값 조회다.**

### 실습 — 그대로 실행하기

```python
prices = {'사과': 1000, '배': 2000}
price = prices['사과']
print(price)  # 예상 출력: 1000
print(type(prices).__name__)  # 예상 출력: dict
print(type(price).__name__)  # 예상 출력: int
print(prices)  # 예상 출력: {'사과': 1000, '배': 2000}
empty = {}
print(empty)  # 예상 출력: {}
```

`price = prices['사과']`의 오른쪽을 먼저 계산합니다.

| 단계 | 값 | 타입 |
| --- | --- | --- |
| 찾는 키 | '사과' | str |
| 해당 키에 저장된 값 | 1000 | int |
| price에 들어가는 값 | 1000 | int |

`prices[1000]`은 가격 1000의 물건을 역으로 찾는 표현이 아닙니다. **1000이라는 키**를 찾습니다. 이 데이터에는 그 키가 없으므로 `KeyError`입니다. 숫자도 키로 사용할 수는 있지만, 이 교재의 예시는 문자열 키로 통일합니다.

## 2. 변수로 조회할 때 따옴표를 붙이지 않는다

### 개념

`name = '사과'`라면 `prices[name]`은 `prices['사과']`와 같습니다. 변수 name을 평가한 결과가 키가 됩니다.

반면 `prices['name']`은 글자 그대로 `'name'`이라는 키를 찾습니다. 함수 인수로 받은 이름을 조회할 때 이 차이가 중요합니다.

### 실습 — 그대로 실행하기

```python
prices = {'사과': 1000, 'name': 9999}
name = '사과'
print(name)  # 예상 출력: 사과
print(prices[name])  # 예상 출력: 1000
print(prices['name'])  # 예상 출력: 9999

# 함수도 같은 조회를 합니다. return은 찾은 값을 호출한 곳으로 돌려줍니다.
def lookup_model(prices, name):
    return prices[name]

print(lookup_model({'배': 2000}, '배'))  # 예상 출력: 2000
```

| 표현 | 딕셔너리에서 찾는 키 | 결과 |
| --- | --- | --- |
| prices[name] | name에 담긴 '사과' | 1000 |
| prices['name'] | 문자열 'name' | 9999 |

함수를 호출하면 인수 `'배'`가 함수 안의 name에 들어갑니다. 그래서 `return prices[name]`은 이 호출에서 `return prices['배']`처럼 동작합니다.

이 단계에서는 **찾는 키가 존재한다고 가정**합니다. 없는 키 처리는 4~5장에서 배웁니다.

## 3. 대입하면 추가하거나 수정한다

### 개념

조회는 `d[key]`, 저장은 `d[key] = value`입니다. 같은 대괄호 표현이라도 `=`의 왼쪽에 있으면 저장 대상입니다.

키가 없으면 새 항목을 추가하고, 이미 있으면 그 키의 값만 바꿉니다. 다른 키는 남습니다. 대입은 원본을 바꾸는 문장이며 메서드처럼 반환값을 돌려주지 않습니다.

### 실습 — 그대로 실행하기

```python
stock = {}
stock['사과'] = 3
print(stock)  # 예상 출력: {'사과': 3}
stock['배'] = 2
print(stock)  # 예상 출력: {'사과': 3, '배': 2}
stock['사과'] = 5
print(stock)  # 예상 출력: {'사과': 5, '배': 2}
print(stock['사과'])  # 예상 출력: 5
```

| 실행한 문장 | 의미 |
| --- | --- |
| stock = {} | 새 빈 딕셔너리를 만들어 stock에 연결 |
| stock['사과'] = 3 | 새 키 '사과' 추가 |
| stock['배'] = 2 | 새 키 '배' 추가 |
| stock['사과'] = 5 | 기존 '사과'의 값만 5로 변경 |

같은 키의 항목 두 개가 생기는 것이 아닙니다. 앞의 3이 5로 바뀝니다.

`stock = {}`를 다시 실행하면 stock은 새 빈 딕셔너리를 가리킵니다. 기존 내용을 계속 쓸 때 이 문장을 반복하지 마세요.

여기까지 읽었다면 **연습 1~2번**을 풀 준비가 되었습니다. 바로 풀지 않고 다음 장까지 이어 읽어도 됩니다.

## 4. in은 키가 있는지 검사한다

### 개념

`key in d`는 키가 있으면 `True`, 없으면 `False`를 반환합니다. 타입은 bool이고 원본은 바뀌지 않습니다. `key not in d`는 반대입니다.

**이건 코테에서 자주 쓰니 챙겨라: 값이 0이어도 키는 존재할 수 있다.** `if d[key]`로 존재를 판단하면 0인 값을 놓치고, 없는 키에서는 오류가 납니다.

### 실습 — 그대로 실행하기

```python
stock = {'사과': 0, '배': 2}
print('사과' in stock)  # 예상 출력: True
print('포도' in stock)  # 예상 출력: False
print('포도' not in stock)  # 예상 출력: True
print(2 in stock)  # 예상 출력: False
print('사과' in {})  # 예상 출력: False

name = '포도'
if name in stock:
    result = stock[name]
else:
    result = '등록되지 않음'
print(result)  # 예상 출력: 등록되지 않음
```

`'사과' in stock`은 등록 여부를 묻고, `stock['사과']`는 재고 수 0을 읽습니다. 서로 다른 질문입니다.

| 표현 | 확인하는 것 | 결과 |
| --- | --- | --- |
| '사과' in stock | '사과'라는 키가 있나? | True |
| stock['사과'] | 그 키의 값은? | 0 |
| 2 in stock | 숫자 2라는 키가 있나? | False |

마지막 실습에서는 `'포도'`가 없으므로 else로 이동합니다. 위험한 직접 조회를 실행하지 않아 `KeyError`가 나지 않습니다.

## 5. get은 없는 키에 사용할 기본값을 정한다

### 개념

`d.get(key, default)`는 키가 있으면 저장된 값, 없으면 default를 반환합니다. 기본값을 생략한 `d.get(key)`는 없는 키에서 `None`을 반환합니다.

**이건 코테에서 자주 쓰니 챙겨라: `get(키, 기본값)`.** 별도의 if 없이 안전하게 읽을 수 있습니다. 저장된 값이나 기본값의 타입을 그대로 반환하고 원본을 변경하지 않습니다.

### 실습 — 그대로 실행하기

```python
scores = {'민수': 80, '지수': 0}
print(scores.get('민수', -1))  # 예상 출력: 80
print(scores.get('지수', -1))  # 예상 출력: 0
print(scores.get('영희', -1))  # 예상 출력: -1
print(scores.get('영희'))  # 예상 출력: None
print({}.get('민수', -1))  # 예상 출력: -1
print(scores)  # 예상 출력: {'민수': 80, '지수': 0}
```

`scores.get('영희', -1)`은 -1을 돌려줄 뿐, `'영희': -1`을 저장하지 않습니다.

| 표현 | 키가 존재할 때 | 키가 없을 때 |
| --- | --- | --- |
| d[key] | 저장된 값 | KeyError |
| key in d | True | False |
| d.get(key, -1) | 저장된 값 | -1 |

지수의 점수 0은 그대로 반환됩니다. `scores.get('지수') or -1`처럼 쓰면 0도 거짓으로 처리해 -1이 되므로, 기본값 인수를 쓰세요.

지금까지 익힌 표현으로 **연습 3~4번**을 풀 수 있습니다.

## 6. 반복문은 기본적으로 키를 꺼낸다

### 개념

`for name in stock:`에서 name에는 **키**가 하나씩 들어갑니다. 숫자 인덱스나 값이 자동으로 들어오는 것이 아닙니다.

값만 필요하면 `stock.values()`를 사용합니다. 인수 없이 호출하며 값들을 순회할 수 있는 `dict_values` 뷰를 반환합니다. 원본을 바꾸지 않습니다. 전체 값의 합은 `sum(stock.values())`가 짧은 정석입니다.

### 실습 — 그대로 실행하기

```python
stock = {'사과': 3, '배': 2}
print(sum(stock.values()))  # 예상 출력: 5

# 값만 꺼내면 [3, 2], 키를 꺼내면 ['사과', '배']입니다.
print(list(stock.values()))  # 예상 출력: [3, 2]
print(list(stock.keys()))  # 예상 출력: ['사과', '배']

# 위의 합계를 일반 for문으로 풀어 쓰면 다음과 같습니다.
total = 0
for name in stock:
    count = stock[name]
    total += count
print(total)  # 예상 출력: 5
print(sum({}.values()))  # 예상 출력: 0
```

| 반복 | name (str) | stock[name] (int) | 누적 total |
| --- | --- | --- | --- |
| 첫 번째 | '사과' | 3 | 3 |
| 두 번째 | '배' | 2 | 5 |

`sum(stock)`은 키를 더하려고 하므로 여기서는 맞지 않습니다.

`keys()`도 인수 없이 호출하며 키의 뷰(dict_keys)를 반환합니다. 두 뷰는 원본의 변경을 반영합니다. `list(뷰)`는 그 시점의 내용으로 새 리스트를 만듭니다. 빈 딕셔너리의 뷰는 비어 있고 반복하지 않습니다. `sum`은 빈 입력에서 기본 합 0을 반환합니다.

`2 in stock.values()`처럼 값 존재도 확인할 수 있지만 키 조회와 달리 값을 차례로 살펴봅니다. **연습 5번**과 연결됩니다.

## 7. items는 키와 값을 함께 꺼낸다

### 개념

`for name, score in scores.items():`는 이름과 점수를 한 쌍씩 받습니다. `items()`는 인수 없이 호출하고 `(키, 값)` 튜플들의 뷰(dict_items)를 반환합니다. 원본은 바뀌지 않습니다.

**이건 코테에서 자주 쓰니 챙겨라: 키와 값이 둘 다 필요하면 `items()`.** 이번에는 점수로 판단하고 이름을 모읍니다.

### 실습 — 그대로 실행하기

```python
scores = {'민수': 80, '지수': 59, '영희': 60}
passed = []
for name, score in scores.items():
    if score >= 60:
        passed.append(name)
print(passed)  # 예상 출력: ['민수', '영희']
print(list(scores.items()))  # 예상 출력: [('민수', 80), ('지수', 59), ('영희', 60)]
print(list({}.items()))  # 예상 출력: []
```

| 받은 튜플 | name | score | 조건 | passed |
| --- | --- | --- | --- | --- |
| ('민수', 80) | '민수' | 80 | 참 | ['민수'] |
| ('지수', 59) | '지수' | 59 | 거짓 | ['민수'] |
| ('영희', 60) | '영희' | 60 | 참 | ['민수', '영희'] |

`name, score`가 튜플의 두 값을 각각 받습니다. 순회 순서는 삽입 순서입니다(Python 3.7 이상). `items()`도 원본 변경을 반영하는 뷰입니다. 순회하면서 키를 추가·삭제하지 마세요.

빈 딕셔너리나 합격자가 없는 경우 passed는 빈 리스트 그대로입니다. **연습 6번**과 연결됩니다.

## 8. get으로 등장 횟수를 누적한다

### 개념

새 글자가 나오면 1부터 세고, 이미 나온 글자면 기존 개수에 1을 더합니다.

정석 표현은 `counts[ch] = counts.get(ch, 0) + 1`입니다. 오른쪽에서 **기존 개수 읽기 → 1 더하기**를 하고, 왼쪽에 저장합니다. `get`만 호출하면 저장되지 않습니다.

### 실습 — 그대로 실행하기

```python
text = 'aba'
counts = {}
for ch in text:
    counts[ch] = counts.get(ch, 0) + 1
print(counts)  # 예상 출력: {'a': 2, 'b': 1}

# 같은 과정을 중간 변수로 풀어 쓰기
counts = {}
for ch in text:
    previous = counts.get(ch, 0)
    updated = previous + 1
    counts[ch] = updated
print(counts)  # 예상 출력: {'a': 2, 'b': 1}
```

| ch | previous | updated | 저장 후 counts |
| --- | --- | --- | --- |
| 'a' | 0 | 1 | {'a': 1} |
| 'b' | 0 | 1 | {'a': 1, 'b': 1} |
| 'a' | 1 | 2 | {'a': 2, 'b': 1} |

counts는 **반복문 밖**에서 한 번 만듭니다. 반복문 안에 `counts = {}`가 있으면 매번 새 기록을 시작합니다. 처음 보는 키에 `counts[ch] += 1`을 쓰면 읽을 값이 없어서 KeyError입니다.

빈 문자열이면 반복하지 않아 `{}`가 남습니다. 실제로 빈도표만 필요할 때는 더 짧은 [Counter](counter.md)도 챙기세요. 여기서는 먼저 get과 대입이 누적을 만드는 과정을 익힙니다. **연습 7번**과 연결됩니다.

## 9. 횟수 대신 수량을 더할 수도 있다

### 개념

앞에서는 등장할 때마다 1을 더했습니다. 주문 수량을 합칠 때는 1 대신 quantity를 더합니다.

`orders`는 딕셔너리가 아니라 **튜플들이 들어 있는 리스트**입니다. 그래서 `for name, quantity in orders:`로 바로 순회합니다. `.items()`를 붙이지 않습니다.

### 실습 — 그대로 실행하기

```python
orders = [('사과', 2), ('배', 1), ('사과', 3)]
totals = {}
for name, quantity in orders:
    totals[name] = totals.get(name, 0) + quantity
print(totals)  # 예상 출력: {'사과': 5, '배': 1}
```

| name | quantity | 기존 합계 | 새 합계 |
| --- | --- | --- | --- |
| '사과' | 2 | 0 | 2 |
| '배' | 1 | 0 | 1 |
| '사과' | 3 | 2 | 5 |

`totals[name] = quantity`만 쓰면 사과 값은 마지막 주문 수량 3이 됩니다. 누적하려면 기존 값에 더해야 합니다.

**횟수는 +1, 수량 합계는 +quantity.** 빈 주문 리스트의 결과는 `{}`입니다. 입력 orders는 바뀌지 않고 새 딕셔너리를 만듭니다. **연습 8번**과 연결됩니다.

## 10. 딕셔너리에 위치를 저장한다

### 개념

딕셔너리 값은 개수일 필요가 없습니다. 글자를 키로, 마지막 위치를 값으로 저장할 수 있습니다.

`enumerate(text)`를 순회하면 `(인덱스, 글자)`를 받습니다. idx는 int, ch는 str입니다. enumerate는 이런 쌍을 순서대로 제공하는 객체를 반환하고 원본을 바꾸지 않습니다. **이건 코테에서 자주 쓰니 챙겨라: 위치와 원소가 함께 필요하면 enumerate.**

### 실습 — 그대로 실행하기

```python
text = 'aba'
last_seen = {}
for idx, ch in enumerate(text):
    last_seen[ch] = idx
print(last_seen)  # 예상 출력: {'a': 2, 'b': 1}
print(last_seen['a'])  # 예상 출력: 2
print(list(enumerate(text)))  # 예상 출력: [(0, 'a'), (1, 'b'), (2, 'a')]
```

| idx (위치) | ch (글자) | 저장 문장 | 결과 |
| --- | --- | --- | --- |
| 0 | 'a' | last_seen['a'] = 0 | {'a': 0} |
| 1 | 'b' | last_seen['b'] = 1 | {'a': 0, 'b': 1} |
| 2 | 'a' | last_seen['a'] = 2 | {'a': 2, 'b': 1} |

3장에서 배운 **같은 키 덮어쓰기** 덕분에 마지막 위치가 남습니다. 딕셔너리가 위치를 자동으로 알아내는 것이 아니라 우리가 idx를 넣습니다.

빈 문자열이면 enumerate도 아무 쌍을 제공하지 않아 결과는 `{}`입니다. 상세 문법은 [enumerate 문서](../04-반복과-변환/enumerate.ipynb)를 참고하세요. **연습 9번**과 연결됩니다.

## 11. 이전 값을 사용한 뒤에 갱신한다

### 개념

이제 각 글자에서 왼쪽의 가장 가까운 같은 글자까지 거리를 구합니다. 이전에 없으면 -1입니다.

현재 위치가 2이고 이전 a의 위치가 0이면 거리는 `2 - 0 = 2`입니다. **이전 위치 읽기 → 거리 계산 → 현재 위치 저장** 순서로 처리합니다.

### 실습 — 그대로 실행하기

```python
text = 'abaca'
answer = []
last_seen = {}
for idx, ch in enumerate(text):
    if ch in last_seen:
        previous = last_seen[ch]
        distance = idx - previous
        answer.append(distance)
    else:
        answer.append(-1)
    last_seen[ch] = idx
print(answer)  # 예상 출력: [-1, -1, 2, -1, 2]
print(last_seen)  # 예상 출력: {'a': 4, 'b': 1, 'c': 3}
```

| idx | ch | 갱신 전 이전 위치 | 추가할 값 | 실행 끝의 last_seen |
| --- | --- | --- | --- | --- |
| 0 | 'a' | 없음 | -1 | {'a': 0} |
| 1 | 'b' | 없음 | -1 | {'a': 0, 'b': 1} |
| 2 | 'a' | 0 | 2 | {'a': 2, 'b': 1} |
| 3 | 'c' | 없음 | -1 | {'a': 2, 'b': 1, 'c': 3} |
| 4 | 'a' | 2 | 2 | {'a': 4, 'b': 1, 'c': 3} |

`last_seen[ch] = idx`는 if/else **바깥, for 안쪽**에 있습니다. 처음 본 글자도, 다시 본 글자도 현재 위치를 기록해야 하기 때문입니다.

먼저 덮어쓰면 이전 위치가 사라져 현재 위치끼리 빼게 됩니다. 또 `if last_seen.get(ch)`는 이전 위치가 0일 때 거짓이 되므로 존재 확인에는 `in`을 씁니다.

빈 문자열이면 answer는 `[]`입니다. **연습 10번**과 연결됩니다.

| 작업 | 시간 | 결과 포함 추가 공간 |

값을 리스트로 바꾸는 `list(d.values())` 등은 항목 수 m에 대해 시간·공간 O(m)입니다. 앞 실습에서 내용 확인을 위해 사용했지만, 합계만 구할 때는 `sum(d.values())`로 충분합니다.
