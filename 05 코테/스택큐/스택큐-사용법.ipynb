# 스택·큐 사용법

[스택·큐 목차](README.md) · [코테 목차](../README.md)

**스택은 보통 `list`, 큐는 보통 `deque`로 구현한다.** 코테에서 자주 쓰니 `append`, `pop`, `popleft`를 챙기자.

- 스택·큐: 값을 넣고 꺼내는 **규칙**.
- `list`·`deque`: 그 규칙을 구현할 때 고르는 **실제 자료형**.
- `deque`도 `append()`와 `pop()`을 사용하면 스택처럼 쓸 수 있다.


## 1. 스택: 리스트로 뒤에 넣고 뒤에서 꺼내기

`append(값)`은 원본 뒤에 값을 추가하고 `None`을 반환한다. `pop()`은 원본의 마지막 값을 제거하고 **꺼낸 값**을 반환한다. 반환 타입은 꺼낸 값의 타입이다.

`stack[-1]`은 마지막 값을 **확인만** 한다. 제거하지 않는다.


````python
stack = []
stack.append(10)
stack.append(20)
stack.append(30)
print(stack)          # [10, 20, 30]
print(stack[-1])      # 30: 확인만 함

removed = stack.pop()
print(removed)        # 30: 꺼낸 값 (int)
print(stack)          # [10, 20]: 원본도 변경됨
````

## 2. 큐: deque로 뒤에 넣고 앞에서 꺼내기

`from collections import deque`로 가져온다. Python에 포함된 **표준 라이브러리**라 별도 설치는 필요 없다.

`deque()`는 빈 deque, `deque(반복 가능한 값)`은 해당 값들을 담은 **새 deque**를 만든다. 리스트를 전달해도 결과는 리스트가 아니다.

`append(값)`은 뒤에 추가하고 `None`을 반환한다. **`popleft()`**는 앞에서 제거하고 꺼낸 값을 반환한다. `leftpop`이 아니다.


````python
from collections import deque

numbers = [10, 20, 30]
queue = deque(numbers)
print(type(numbers))  # <class 'list'>
print(type(queue))    # <class 'collections.deque'>

queue.append(40)
print(queue[0])       # 10: 맨 앞 확인, 제거하지 않음
first = queue.popleft()
print(first)          # 10
print(list(queue))    # [20, 30, 40]: 보기 위해 리스트로 변환
print(numbers)        # [10, 20, 30]: 원래 리스트는 그대로
````

## 3. 필요한 메서드만 찾아보기

아래 추가·제거 메서드는 모두 원본을 바꾼다. 조회는 원본을 바꾸지 않는다.

| 목적 | list | deque | 반환값 |
| --- | --- | --- | --- |
| 뒤에 한 값 추가 | `a.append(x)` | `q.append(x)` | `None` |
| 뒤에서 한 값 제거 | `a.pop()` | `q.pop()` | 꺼낸 값 |
| 앞에서 한 값 제거 | `a.pop(0)` | `q.popleft()` | 꺼낸 값 |
| 앞에 한 값 추가 | `a.insert(0, x)` | `q.appendleft(x)` | `None` |
| 앞 / 뒤 값 확인 | `a[0]` / `a[-1]` | `q[0]` / `q[-1]` | 해당 값 |
| 개수 확인 | `len(a)` | `len(q)` | 정수 |

**빈 상태에서 `pop()`, `popleft()`, `[0]`, `[-1]`을 쓰면 `IndexError`가 난다.**

`queue = queue.append(10)`으로 쓰면 안 된다. `append()`의 반환값인 `None`이 변수에 저장된다. 그냥 `queue.append(10)`으로 호출한다.


````python
from collections import deque

queue = deque([20, 30])
queue.appendleft(10)
queue.append(40)
print(list(queue))      # [10, 20, 30, 40]
print(queue.pop())      # 40
print(queue.popleft())  # 10
print(list(queue))      # [20, 30]
````

## 4. 왜 큐에서는 리스트 대신 deque를 쓰나?

리스트의 `pop(0)`도 동작한다. 하지만 뒤에 남은 원소들의 참조를 한 칸씩 앞으로 당겨야 한다. **크기순 재정렬이 아니라 위치를 당기는 작업**이다.

```text
list: [10, 20, 30] → pop(0) → [20, 30]
                         20과 30을 앞으로 당김

deque: popleft()로 앞 원소를 꺼낼 때 나머지를 전부 당기지 않음
```

`n`은 현재 원소 개수다.

| 동작 | list | deque |
| --- | --- | --- |
| 뒤에 추가 | 분할 상환 O(1) | O(1) |
| 뒤에서 제거 | 분할 상환 O(1) | O(1) |
| 앞에 추가·앞에서 제거 | O(n) | O(1) |
| 중간 인덱스 접근 | O(1) | 최악 O(n), 양 끝은 O(1) |
| 슬라이싱 | 가능 | 직접 사용 불가 |

분할 상환 O(1)은 여러 번 수행했을 때 한 번당 비용이 평균적으로 일정하다는 뜻이다. 리스트는 가끔 내부 공간을 조정한다.


## 5. 큐가 빌 때까지 꺼내는 기본형

빈 deque는 거짓으로 판단된다. `while queue:`는 값이 남아 있을 때만 실행되므로, 안에서 안전하게 꺼낼 수 있다.


````python
from collections import deque

queue = deque([10, 20, 30])
while queue:
    current = queue.popleft()
    print(current)
# 10
# 20
# 30

print(bool(queue))  # False
print(len(queue))   # 0
````
