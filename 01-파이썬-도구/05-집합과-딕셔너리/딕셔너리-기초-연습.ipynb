### 해설 — 직접 시도한 다음 읽기

- `items()`는 인수 없이 (키, 값) 쌍의 뷰(dict_items)를 반환합니다. `keys()`는 키 뷰, `values()`는 값 뷰입니다. 모두 조회이며 원본을 바꾸지 않습니다.
- 그냥 `for name in scores`로 순회하면 키만 나옵니다. 점수 대신 이름을 비교하지 마세요.
- 빈 딕셔너리에서는 반복하지 않습니다. 뷰는 원본 변경을 반영하고 `list(뷰)`는 그 시점의 리스트입니다.

| name | score | score >= 60 | answer |
| --- | --- | --- | --- |
| 민수 | 80 | True | ['민수'] |
| 지수 | 59 | False | ['민수'] |
| 영희 | 60 | True | ['민수', '영희'] |

항목 수 m, 합격자 수 k에 대해 시간 O(m), 공간 O(k).

가져갈 것: **키·값을 함께 사용할 때 `items()`.**

### 모범 코드

```python
def solution_06_model(scores, cutoff):
    answer = []
    for name, score in scores.items():
        if score >= cutoff:
            answer.append(name)
    return answer

print(solution_06_model({'민수': 80, '지수': 59, '영희': 60}, 60))  # 예상 출력: ['민수', '영희']
print(solution_06_model({'민수': 59}, 60))  # 예상 출력: []
print(solution_06_model({}, 60))  # 예상 출력: []
```

## 7. 과일별 개수 세기

### 문제

과일 이름(str) 리스트 `fruits`를 받아 이름 → 등장 횟수(int)인 새 딕셔너리를 반환하세요. 빈 리스트는 `{}`입니다. 이번에는 `get`으로 누적해보세요.

작은 문법: `counts = {}`에서 `counts['x'] = counts.get('x', 0) + 1`을 실행하면 `{'x': 1}`입니다.

| 입력 인수 | 반환값 |
| --- | --- |
| `(['사과', '배', '사과'])` | `{'사과': 2, '배': 1}` |
| `(['배', '배', '배'])` | `{'배': 3}` |
| `([])` | `{}` |

<details>
<summary>막히면 힌트 보기</summary>

누적할 딕셔너리를 반복문 밖에서 만들고, 현재 개수에 1을 더해 저장하세요.

</details>

### 다시 풀기

```python
def solution_07(fruits):
    # 여기에 직접 작성하세요.
    stock = {}
    
    for f in fruits:
        stock[f] = stock.get(f,0) +1

    return stock
```

```python
print(solution_07(['사과', '배', '사과']))  # 예상 출력: {'사과': 2, '배': 1}
print(solution_07(['배', '배', '배']))  # 예상 출력: {'배': 3}
print(solution_07([]))  # 예상 출력: {}
```

### 해설 — 직접 시도한 다음 읽기

- 없는 키를 `counts[fruit] += 1`로 바로 읽으면 KeyError입니다. get으로 시작값 0을 사용합니다.
- 반복문 안에서 `counts = {}`를 실행하면 앞에서 센 기록을 잃습니다.

| fruit | 기존 개수(get) | +1 후 counts |
| --- | --- | --- |
| 사과 | 0 | {'사과': 1} |
| 배 | 0 | {'사과': 1, '배': 1} |
| 사과 | 1 | {'사과': 2, '배': 1} |

`counts.get(fruit, 0) + 1`은 읽기·계산이고, 왼쪽 `counts[fruit] =`가 결과를 저장합니다.

입력 길이 n, 서로 다른 과일 수 u에 대해 시간 O(n), 공간 O(u). 단순 빈도 계산의 실전 도구는 [Counter](counter.md)입니다. 이번 문제는 누적 문법 연습이며, Counter의 상세 사용법은 해당 문서에서 이어갑니다.

가져갈 것: **`d[key] = d.get(key, 0) + 1`.**

### 모범 코드

```python
def solution_07_model(fruits):
    counts = {}
    for fruit in fruits:
        counts[fruit] = counts.get(fruit, 0) + 1
    return counts

print(solution_07_model(['사과', '배', '사과']))  # 예상 출력: {'사과': 2, '배': 1}
print(solution_07_model(['배', '배', '배']))  # 예상 출력: {'배': 3}
print(solution_07_model([]))  # 예상 출력: {}
```

## 8. 물건별 주문 수량 합치기

### 문제

`orders`는 `(물건 이름 str, 주문 수량 int)` 튜플의 리스트입니다. 수량은 0~100이고 이름이 반복될 수 있습니다. 물건별 총수량을 새 딕셔너리로 반환하세요. 0개 주문된 물건도 키에 포함하세요.

7번에서 매번 더한 1을 이번에는 주문 수량으로 바꾸는 연습입니다.

| 입력 인수 | 반환값 |
| --- | --- |
| `([('사과', 2), ('배', 1), ('사과', 3)])` | `{'사과': 5, '배': 1}` |
| `([('사과', 0)])` | `{'사과': 0}` |
| `([])` | `{}` |

<details>
<summary>막히면 힌트 보기</summary>

for name, quantity in orders로 두 값을 받은 뒤 기존 합계에 quantity를 더하세요.

</details>

### 다시 풀기

```python
def solution_08(orders):
    # 여기에 직접 작성하세요.
    total = {}

    for name, q in orders:
        total[name] = total.get(name, 0) + q
    return total
```

```python
print(solution_08([('사과', 2), ('배', 1), ('사과', 3)]))  # 예상 출력: {'사과': 5, '배': 1}
print(solution_08([('사과', 0)]))  # 예상 출력: {'사과': 0}
print(solution_08([]))  # 예상 출력: {}
```

### 해설 — 직접 시도한 다음 읽기

- `totals[name] = quantity`만 쓰면 마지막 주문 수량만 남습니다.
- 개수를 세는 문제와 합계를 구하는 문제는 더하는 값이 다릅니다.

| name | quantity | 기존 합계 | 새 합계 |
| --- | --- | --- | --- |
| 사과 | 2 | 0 | 2 |
| 배 | 1 | 0 | 1 |
| 사과 | 3 | 2 | 5 |

orders는 리스트이므로 `.items()`를 붙이지 않습니다. 각 원소인 튜플을 두 변수로 나눠 받습니다. 빈 리스트면 반복하지 않아 `{}`를 반환합니다.

주문 수 n, 물건 종류 u에 대해 시간 O(n), 공간 O(u).

가져갈 것: **누적 대상이 횟수인지 수량인지 구분.**

### 모범 코드

```python
def solution_08_model(orders):
    totals = {}
    for name, quantity in orders:
        totals[name] = totals.get(name, 0) + quantity
    return totals

print(solution_08_model([('사과', 2), ('배', 1), ('사과', 3)]))  # 예상 출력: {'사과': 5, '배': 1}
print(solution_08_model([('사과', 0)]))  # 예상 출력: {'사과': 0}
print(solution_08_model([]))  # 예상 출력: {}
```

### 해설 — 직접 시도한 다음 읽기

- idx는 위치(int), ch는 글자(str)입니다. 딕셔너리가 위치를 계산하는 것이 아니라 우리가 idx를 저장합니다.
- `last_seen[idx] = ch`는 요구한 대응 방향과 반대입니다.

| idx | ch | 저장 후 last_seen |
| --- | --- | --- |
| 0 | a | {'a': 0} |
| 1 | b | {'a': 0, 'b': 1} |
| 2 | a | {'a': 2, 'b': 1} |

enumerate는 iterable을 받아 (인덱스, 원소)를 순서대로 제공하는 enumerate 객체를 반환합니다. 원본을 바꾸지 않고 빈 입력에서는 아무 쌍도 나오지 않습니다.

가져갈 것: **글자 → 위치 대응과 같은 키 덮어쓰기.**

### 모범 코드

```python
def solution_09_model(text):
    last_seen = {}
    for idx, ch in enumerate(text):
        last_seen[ch] = idx
    return last_seen

print(solution_09_model('abaca'))  # 예상 출력: {'a': 4, 'b': 1, 'c': 3}
print(solution_09_model('a'))  # 예상 출력: {'a': 0}
print(solution_09_model(''))  # 예상 출력: {}
```

### 해설 — 직접 시도한 다음 읽기

- 현재 위치를 먼저 저장하면 `idx - last_seen[ch]`가 항상 0이 됩니다.
- `if last_seen.get(ch)`로 존재를 검사하면 이전 위치 0을 놓칩니다. `ch in last_seen`을 사용하세요.
- 기록 갱신을 else 안에만 두면 반복된 글자의 최신 위치가 저장되지 않습니다.

| idx | ch | 갱신 전 이전 위치 | 넣을 거리 | 갱신 후 a 위치 |
| --- | --- | --- | --- | --- |
| 0 | a | 없음 | -1 | 0 |
| 1 | b | 없음 | -1 | 0 |
| 2 | a | 0 | 2 - 0 = 2 | 2 |
| 3 | c | 없음 | -1 | 2 |
| 4 | a | 2 | 4 - 2 = 2 | 4 |

문자열 길이 n, 서로 다른 글자 수 u에 대해 시간 O(n), 반환 리스트 포함 공간 O(n + u). 기록용 딕셔너리만 보면 O(u)입니다.

가져갈 것: **이전 기록 조회 → 계산 → 현재 기록 갱신.**

### 모범 코드

```python
def solution_10_model(text):
    answer = []
    last_seen = {}
    for idx, ch in enumerate(text):
        if ch in last_seen:
            previous = last_seen[ch]
            answer.append(idx - previous)
        else:
            answer.append(-1)
        last_seen[ch] = idx
    return answer

print(solution_10_model('abaca'))  # 예상 출력: [-1, -1, 2, -1, 2]
print(solution_10_model('aaa'))  # 예상 출력: [-1, 1, 1]
print(solution_10_model(''))  # 예상 출력: []
```
