## 문제 설명

## 해설

### 제공한 풀이 진단

`zip`으로 진도와 속도를 묶고, `math.ceil`로 완료까지 걸리는 일수를 구한 뒤 `deque`에서 앞부터 꺼내는 로직은 맞다. `while que and que[0] <= depoly_day`로 같은 날 함께 배포할 기능을 세는 조건도 정확하다. 플랫폼 제출 통과 여부는 확인하지 않았다.

논리 오류는 없으며 모범 코드에서는 다음 두 곳만 다듬었다.

- `day = 0`은 곧바로 계산 결과로 덮어쓰므로 생략한다.
- `depoly_day`는 `deploy_day`로 철자를 고친다. 기존 이름도 일관되게 사용했으므로 실행 오류는 아니다.

### 핵심 아이디어

1. 각 기능의 완료 일수를 계산해 **원래 순서대로** 큐에 넣는다. 완료가 빠르다고 정렬하면 배포 순서가 바뀐다.
2. 맨 앞 값을 꺼내 이번 묶음의 `deploy_day`로 정하고 `count = 1`로 시작한다.
3. 다음 기능의 완료일이 `deploy_day` 이하면 꺼내며 개수를 늘린다.
4. 더 오래 걸리는 기능을 만나면 이번 개수를 저장하고, 다음 묶음을 시작한다.

묶음 안에서는 `deploy_day`를 바꾸지 않는다. `[7, 3, 6]`은 모두 7일째 배포된다. 중간에 기준을 3으로 바꾸면 6일째 완료되는 기능을 잘못 분리한다. `[7, 9, 3]`에서는 9일 걸리는 기능을 건너뛸 수 없으므로 `[1, 2]`가 된다.

### 값과 타입 추적: 첫 번째 예제

| progress (`int`) | speed (`int`) | amount (`int`) | amount / speed (`float`) | day (`int`) |
| --- | --- | --- | --- | --- |
| 93 | 1 | 7 | 7.0 | 7 |
| 30 | 30 | 70 | 약 2.333 | 3 |
| 55 | 5 | 45 | 9.0 | 9 |

계산을 마치면 `que = deque([7, 3, 9])`다. 큐에 들어 있는 수는 진도나 인덱스가 아니라 **오늘부터 완료까지 걸리는 일수**다.

| 동작 | deploy_day | count | 남은 que | answer |
| --- | --- | --- | --- | --- |
| 맨 앞 7을 꺼내 첫 배포 시작 | 7 | 1 | `[3, 9]` | `[]` |
| `3 <= 7`이므로 함께 배포 | 7 | 2 | `[9]` | `[]` |
| `9 > 7`이므로 묶음 종료 | 7 | 2 | `[9]` | `[2]` |
| 9를 꺼내 다음 배포, 큐가 비어 종료 | 9 | 1 | `[]` | `[2, 1]` |

두 번째 예제의 완료 일수는 `[5, 10, 1, 1, 20, 1]`이고, `[5] / [10, 1, 1] / [20, 1]`로 묶인다.

### Python 도구와 조건

이건 코테에서 자주 쓰니 챙겨라: **올림으로 필요한 일수 계산하기**, **`deque.popleft()`로 앞에서 꺼내기**, **`que and ...`로 빈 큐 보호하기**.

- `zip(progresses, speeds)`는 같은 위치의 값을 `(progress, speed)` 튜플로 제공하는 이터레이터다. 입력을 바꾸지 않으며, 짧은 입력이 끝나면 멈춘다. 한쪽이 비어 있으면 반복하지 않는다. [zip 상세](../../01-파이썬-도구/04-반복과-변환/zip-reversed.md)
- `math.ceil(amount / speed)`는 나눗셈 결과를 올림한 `int`를 반환한다. 예를 들어 `math.ceil(70 / 30)`은 `3`, `math.ceil(7 / 1)`은 `7`이다. 숫자 하나를 받으며 원본을 변경하지 않는다. 0은 0이고, 인수를 생략하면 `TypeError`다. 이 문제에서 `speed`는 양수라 0으로 나누지 않는다.
- `que.append(day)`는 큐 뒤에 정수 일수를 추가하고 `None`을 반환한다. `que.popleft()`는 맨 앞 원소를 제거하면서 그 값(`int`)을 반환한다. 둘 다 큐를 변경한다. 빈 큐에서 `popleft()` 또는 `que[0]`을 실행하면 `IndexError`다. [스택·큐 사용법](스택큐-사용법.ipynb)
- `que[0]`은 맨 앞 값을 **조회만** 한다. `while que and que[0] <= deploy_day`는 먼저 큐가 비어 있는지 확인한다. 비었다면 오른쪽 비교를 실행하지 않으므로 안전하다. 순서를 바꾸면 빈 큐 접근을 막지 못한다.

### 조건과 복잡도

- `<=`를 사용하므로 완료일이 같은 기능도 함께 배포한다.
- 하루씩 진도를 갱신할 필요 없이 완료 일수만 계산하면 된다.
- 입력 리스트 `progresses`, `speeds`는 변경하지 않는다. 두 입력이 모두 비어 있다면 이 구현은 `[]`를 반환한다.
- 작업 수를 `n`이라 할 때 시간 O(n), 추가 공간 O(n)이다. 중첩 `while`이지만 각 작업은 큐에 한 번 들어가고 한 번만 나온다. `deque` 양 끝의 추가·제거는 O(1)이다.
- 양의 정수 조건에서는 `(100 - progress + speed - 1) // speed`로도 같은 올림을 계산할 수 있다. `/`의 실수 계산 없이 `int`로 계산하는 대안이다. 현재 입력 범위에서는 원래 `math.ceil` 풀이로 충분하다. [정수 올림 나눗셈](../../01-파이썬-도구/01-숫자/division.md)

## 모범 코드

제공한 풀이의 큐 구조와 `math.ceil` 계산을 그대로 유지했다. 직접 작성할 `solution`과 구분하기 위해 함수 이름은 `solution_model`로 둔다.

```python
from collections import deque
import math


def solution_model(progresses, speeds):
    answer = []
    que = deque()

    for progress, speed in zip(progresses, speeds):
        amount = 100 - progress
        day = math.ceil(amount / speed)
        que.append(day)

    while que:
        deploy_day = que.popleft()
        count = 1

        while que and que[0] <= deploy_day:
            que.popleft()
            count += 1

        answer.append(count)

    return answer


print(solution_model([93, 30, 55], [1, 30, 5]))  # 예상 출력: [2, 1]
print(solution_model([95, 90, 99, 99, 80, 99], [1, 1, 1, 1, 1, 1]))  # 예상 출력: [1, 3, 2]
```

### 가져갈 것

1. 소수 일수가 남아도 하루가 더 필요하므로 완료 일수는 올림한다.
2. 배포 순서를 유지하고, 한 묶음에서는 맨 앞 기능의 배포일을 기준으로 고정한다.
3. `while que and ...`로 안전하게 확인하며, 각 원소를 한 번씩 꺼내면 중첩 반복이어도 O(n)이다.
