## 해설

정리일: 2026-09-25. 대화에서 확인한 풀이와 설명을 기록합니다. 제출 통과나 독립 재풀이 완료를 의미하지 않습니다.

### 답을 보기 전에

1. range의 끝에 n + 1을 넣는 이유는?
2. x가 0 또는 음수여도 같은 풀이가 될까?

1부터 n까지 순회하며 answer.append(x * i)한 첫 풀이가 맞았습니다.

range(1, n + 1)은 1부터 n까지입니다. i가 1, 2, 3일 때 x * i는 x, 2x, 3x입니다.
반복과 append를 리스트 컴프리헨션으로 옮긴 것이며 음수와 0에도 그대로 적용됩니다.
시간 O(n), 반환 리스트 공간 O(n)입니다.

## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(x, n):
    return [x * i for i in range(1, n + 1)]

print(solution_model(2, 5))   # [2, 4, 6, 8, 10]
print(solution_model(-4, 2))  # [-4, -8]
print(solution_model(0, 3))   # [0, 0, 0]
```
