# 올바른 괄호

문제 출처: [프로그래머스 — 올바른 괄호](https://school.programmers.co.kr/learn/courses/30/lessons/12909)

정리일: 2026-09-29

## 문제 설명

괄호가 바르게 짝지어졌다는 것은 `(`로 열렸으면 반드시 짝지어서 `)`로 닫혀야 한다는 뜻입니다.

- `()()` 또는 `(())()`는 올바른 괄호입니다.
- `)()(` 또는 `(()(`는 올바르지 않은 괄호입니다.

`(` 또는 `)`로만 이루어진 문자열 `s`가 주어집니다. 올바른 괄호이면 `True`, 아니면 `False`를 반환하는 `solution` 함수를 완성하세요.

### 제한사항

- 문자열 길이: 1 이상 100,000 이하
- 문자열은 `(`와 `)`로만 구성됩니다.
- 반환값은 `bool`입니다.

### 입출력 예

| s | 반환값 |
| --- | --- |
| `()()` | `True` |
| `(())()` | `True` |
| `)()(` | `False` |
| `(()(` | `False` |


## 다시 풀기

아래 셀에 풀이를 작성하세요. 해설과 모범 코드는 아래쪽에 있습니다.

다시 풀 때는 이 셀을 수정하고 바로 아래 예제 확인 셀을 실행하세요. 모범 코드는 별도 함수 이름을 사용하므로 내 `solution`을 덮어쓰지 않습니다.


````python
def solution(s):
    stk=[]
    for ch in s:
        if ch == "(":
            stk.append(ch)
        else:
            if not stk:
                return False
            stk.pop()
    return not stk
````

````python
print(solution("()()"))    # 예상 출력: True
print(solution("(())()"))  # 예상 출력: True
print(solution(")()("))    # 예상 출력: False
print(solution("(()("))    # 예상 출력: False
````

## 해설

### 핵심 아이디어

**스택에는 아직 닫히지 않은 여는 괄호만 보관한다.**

문자를 전부 넣고 나중에 처리하는 것이 아니라, 문자열을 읽으면서 바로 짝을 맞춘다. 재귀나 별도의 `while`문은 필요 없다.

1. `(`를 만나면 스택에 넣는다.
2. `)`를 만나면 스택에서 여는 괄호 하나를 꺼낸다. 스택이 비어 있으면 짝이 없으므로 바로 `False`다.
3. 끝까지 읽은 뒤 스택이 비어 있어야 `True`다. 남아 있는 `(`는 닫히지 않은 괄호다.

### 이번에 막혔던 세 가지

- **무엇을 쌓을지:** 모든 문자를 쌓으려다 처리 흐름이 막혔다. 미완성인 여는 괄호만 남긴다.
- **값과 인자:** `append('i')`는 문자 `'i'`를 넣는다. 변수의 값을 넣으려면 `append(i)`다. `pop('i')`는 문자열을 인덱스로 전달해서 `TypeError`가 난다. 마지막 값을 꺼낼 때는 `pop()`이다.
- **빈 스택 판정:** `if stack`은 값이 있을 때, `if not stack`은 비어 있을 때 실행된다. 중간의 빈 스택 검사와 마지막의 남은 괄호 검사 모두 필요하다.

### Python 표현

**리스트로 스택을 구현하는 표현은 코테에서 자주 쓰니 챙겨라.** 별도 import는 필요 없다.

| 표현 | 입력·반환 | 원본 변경 / 주의점 |
| --- | --- | --- |
| `stack.append(char)` | 값 하나를 받아 `None` 반환 | 맨 뒤에 추가 |
| `stack.pop()` | 인자 없이 마지막 값 반환; 여기서는 `str` | 맨 뒤 값 제거; 비어 있으면 `IndexError` |
| `stack.pop(index)` | 정수 인덱스 위치의 값 반환 | 해당 위치 제거; 이 풀이에서는 사용하지 않음 |
| `not stack` | 리스트의 빈 여부를 `bool`로 반환 | 변경 없음; `[]`이면 `True` |

`return not stack`은 다음 코드와 같은 뜻이다.

```python
if not stack:
    return True
else:
    return False
```

## 모범 코드

### 1. 스택

직접 완성한 풀이와 같은 로직이다. 복습 셀과 구분하려고 이름을 `solution_stack`으로 썼다. 제출할 때는 함수 이름을 `solution`으로 바꾸면 된다.


````python
def solution_stack(s):
    stack = []

    for char in s:
        if char == '(':
            stack.append(char)
        else:
            if not stack:
                return False
            stack.pop()

    return not stack


print(solution_stack("()()"))    # True
print(solution_stack("(())()"))  # True
print(solution_stack(")()("))    # False
print(solution_stack("(()("))    # False
````

### 실제 값 추적: `(())`

| 읽은 문자 | 동작 | 처리 후 stack |
| --- | --- | --- |
| `(` | 추가 | `['(']` |
| `(` | 추가 | `['(', '(']` |
| `)` | 마지막 여는 괄호 제거 | `['(']` |
| `)` | 마지막 여는 괄호 제거 | `[]` |

마지막에 `not []`는 `True`다.

- `)(`: 첫 `)`에서 스택이 비어 있으므로 즉시 `False`.
- `(()(`: 끝에 `['(', '(']`가 남으므로 `False`.
- `(`: 여는 괄호가 남으므로 `False`.

### 2. 개수 세기

괄호가 한 종류이므로 아직 닫히지 않은 괄호의 개수만 기억해도 된다.
`balance`는 정상적으로 처리하는 동안 스택 풀이의 `len(stack)`에 해당한다.
음수가 되면 여는 괄호 없이 닫는 괄호가 나온 것이므로 즉시 실패한다.


````python
def solution_count(s):
    balance = 0

    for char in s:
        if char == '(':
            balance += 1
        else:
            balance -= 1

        if balance < 0:
            return False

    return balance == 0


print(solution_count("()()"))    # True
print(solution_count("(())()"))  # True
print(solution_count(")()("))    # False
print(solution_count("(()("))    # False
````

### 조건과 복잡도

`n`은 문자열 길이다.

| 풀이 | 시간 | 추가 공간 |
| --- | --- | --- |
| 스택 | O(n) | 최악 O(n): 여는 괄호만 나오는 경우 |
| 개수 세기 | O(n) | O(1) |

리스트 끝의 `append()`는 분할 상환 O(1), 끝의 `pop()`은 O(1)이므로 스택 풀이 전체는 O(n)이다.
문자열을 수정하지 않으며, 빈 문자열은 문제 입력 범위 밖이지만 두 모범 코드는 `True`를 반환한다.
괄호 종류가 여러 개라면 개수만으로 종류와 순서를 확인할 수 없으므로 이 개수 풀이를 그대로 적용할 수 없다.

### 가져갈 것

1. 스택에 무엇을 남길지 먼저 정한다: 아직 닫히지 않은 여는 괄호.
2. 리스트의 `append(값)`과 `pop()`으로 스택을 구현한다.
3. 도중에 짝 없는 닫는 괄호가 나오는 경우와, 마지막에 여는 괄호가 남는 경우를 각각 검사한다.
