### 핵심: 검사와 추가의 시점

바깥 반복문은 숫자를 하나 고르고, 안쪽 반복문은 그 숫자의 문자들을 검사합니다. **안쪽 반복문은 유효성만 검사합니다. 모든 문자 검사가 끝난 뒤 유효한 숫자를 한 번 추가합니다.**

```python
for num in range(l, r + 1):
    valid = True

    for ch in str(num):
        if ch not in "05":
            valid = False
            break

    if valid:
        answer.append(num)
```

`if valid:`는 문자 반복문 밖이지만 숫자 반복문 안입니다. 안쪽 반복문이 끝나면 다음 줄인 `if valid:`로 이동합니다. `break`도 가장 가까운 문자 반복문 하나만 끝냅니다. 다음 숫자는 바깥 반복문에서 계속 처리합니다.

### 실제 실수와 수정 이유

1. `if valid:`를 문자 반복문 안에 두었습니다. 문자 하나만 통과해도 숫자 전체를 추가하므로 아직 검사하지 않은 자릿수의 오류를 놓칩니다.
2. `answer.append(num)`도 문자마다 실행되어 `50`은 두 번, `555`는 세 번 추가되었습니다. 숫자 하나당 검사를 마친 뒤 한 번만 실행해야 합니다.
3. `vaild`는 `valid`의 철자 오타입니다. 같은 철자를 일관되게 사용했으므로 이번 오답 원인은 아닙니다.

### 50과 51로 추적

| 숫자 | 문자 검사 | 기존 코드: 문자마다 추가 | 수정 코드: 검사 후 추가 |
| --- | --- | --- | --- |
| 50 | `"5"` 통과 | 50 추가 | 아직 추가하지 않음 |
| 50 | `"0"` 통과 | 50 또 추가 | 검사 종료 후 50 한 번 추가 |
| 51 | `"5"` 통과 | 51 추가 | 아직 추가하지 않음 |
| 51 | `"1"` 실패 | False로 바꾸고 break, 이미 넣은 51은 남음 | False이므로 추가하지 않음 |

`valid = True`는 숫자를 고를 때마다 다시 설정합니다. 그렇지 않으면 이전 숫자가 실패한 상태를 다음 숫자에도 가져갈 수 있습니다.

### Python 표현 연결

- `range(l, r + 1)`은 l과 r을 모두 포함해 작은 수부터 검사합니다. 별도 정렬은 필요 없습니다.
- `str(num)`으로 정수를 문자열로 바꿔 자릿수를 순회합니다. `num`은 정수, `ch`는 길이 1인 문자열입니다. `answer`에는 정수 `num`을 넣습니다.
- `ch in "05"`는 문자가 0 또는 5인지 검사합니다. 순서를 맞추는 표현이 아니므로 첫 문자가 `"5"`여도 통과합니다.
- 빈 결과의 기본값은 [빈 값과 조건 판단](../01-파이썬-도구/06-문법/truthiness.md), 추가 연산은 [append](../01-파이썬-도구/03-리스트/append.md), 타입 변환은 [int / str](../01-파이썬-도구/04-반복과-변환/type-conversion.md)에서 복습합니다.

## 모범 코드

직접 작성한 `solution`을 덮어쓰지 않도록 모범 함수는 `solution_model`로 구분합니다. 제출할 때는 이름을 `solution`으로 바꾸세요.


````python
def solution_model(l, r):
    answer = []

    for num in range(l, r + 1):
        valid = True

        for ch in str(num):
            if ch not in "05":
                valid = False
                break

        # 모든 문자 검사가 끝난 뒤 숫자를 한 번만 추가한다.
        if valid:
            answer.append(num)

    if not answer:
        return [-1]
    return answer


print(solution_model(5, 555))  # 예상 출력: [5, 50, 55, 500, 505, 550, 555]
print(solution_model(10, 20))  # 예상 출력: [-1]
print(solution_model(50, 51))  # 예상 출력: [50]
````

### 같은 검사를 all로 표현하기

`all(ch in "05" for ch in str(num))`은 모든 문자 검사가 참인지 판단합니다. 앞서 배운 `all()` 풀이와 일반 반복문 풀이는 같은 조건을 검사합니다. 이번 복습의 핵심은 일반 반복문에서 **검사 후 추가하는 위치**를 직접 설명하는 것입니다.

### 경계 조건과 복잡도

- `l = r = 5`이면 `[5]`, `l = r = 51`이면 `[-1]`입니다.
- 양의 정수만 입력되므로 문자열은 비어 있지 않습니다.
- 후보 개수 N = r - l + 1, 최대 자릿수 d, 결과 개수 K라고 하면 시간은 O(Nd), 결과 저장을 포함한 추가 공간은 O(K + d)입니다. 문자열 변환에 O(d) 공간을 사용하며 이 문제에서 d는 최대 7입니다.
