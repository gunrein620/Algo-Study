## 해설

정리일: 2026-09-25. 대화에서 확인한 풀이와 설명을 기록합니다. 제출 통과나 독립 재풀이 완료를 의미하지 않습니다.

### 답을 보기 전에

1. 왜 1은 검사하지 않아도 될까?
2. 왜 n - 1까지 확인하면 반드시 답이 있을까?
3. next로 줄이면 더 빠른 알고리즘이 될까?

작은 수부터 검사한 첫 풀이가 맞았습니다. answer = 0과 마지막 return은 답이 반드시 존재하므로 필요 없습니다.

1로 나눈 나머지는 항상 0입니다. n ≥ 3이므로 n % (n - 1)은 1입니다.
10은 2로 나누면 나머지 0, 3으로 나누면 1이므로 바로 3을 반환합니다.
`return next(x for x in range(2, n) if n % x == 1)`로도 쓸 수 있습니다.
next는 처음 접했고 사용법을 설명받은 상태입니다. 먼저 반복문을 익히고 짧은 표현은 읽을 수 있는 정도로 챙깁니다.
두 방식의 최악 시간은 O(n), 추가 공간은 O(1)입니다.

### 챙길 것

가장 작은 값은 작은 후보부터 검사하고 첫 조건 만족 시 반환한다.

## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(n):
    for x in range(2, n):
        if n % x == 1:
            return x

print(solution_model(10))  # 3
print(solution_model(12))  # 11
```
