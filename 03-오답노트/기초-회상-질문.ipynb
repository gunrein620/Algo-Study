## 해설

1. sorted는 원본을 바꾸지 않고 새 리스트를 반환한다.

2. find는 -1을 반환하지만 index는 ValueError를 발생시킨다.

3. 두 메서드 모두 원본 리스트를 바꾸며 None을 반환한다.

4. lower는 새 문자열을 반환하므로 변경된 값을 쓰려면 저장해야 한다.

5. join에는 문자열 원소가 필요하다. map(str, numbers)로 바꾼다.

6. 조건에 맞을 때 1을 더하면 개수, 원소를 더하면 값의 합이다.

7. 슬라이싱은 시작 위치를 포함하고 끝 위치를 제외한다.

8. //는 결과를 아래쪽 정수로 내린다. 0 쪽으로 자르는 연산이 아니다.

9. isqrt는 제곱근을 내린 정수 후보를 반환한다. 후보의 제곱이 원래 수와 같은지 비교한다.

10. 없는 키를 대괄호로 조회하면 0을 반환한다.

11. 중복 제거 전에 순서 유지가 필요한지 확인한다. dict.fromkeys는 처음 등장한 키의 순서를 유지한다.


## 모범 코드


### 1. `sorted("cab")`의 값과 반환 타입은?


```python
print(sorted('cab'))  # ['a', 'b', 'c']
print(type(sorted('cab')))  # <class 'list'>
```

### 2. 문자열에서 없는 값을 찾으면 find와 index는 각각 어떻게 동작할까?


```python
print('abc'.find('z'))  # -1
try:
    'abc'.index('z')
except ValueError:
    print('ValueError')  # ValueError
```

### 3. 리스트의 sort와 append는 무엇을 반환하고 원본은 어떻게 바뀔까?


```python
arr = [3, 1]
print(arr.sort())  # None
print(arr)  # [1, 3]
print(arr.append(2))  # None
print(arr)  # [1, 3, 2]
```

### 4. s.lower()를 호출하면 s 자체가 바뀔까?


```python
s = 'PyThOn'
print(s.lower())  # python
print(s)  # PyThOn
s = s.lower()
print(s)  # python
```

### 5. 정수 리스트 [1, 2, 3]을 문자열 "123"으로 합치려면?


```python
numbers = [1, 2, 3]
print(''.join(map(str, numbers)))  # 123
```

### 6. sum(1 for ...)과 sum(x for ...)은 무엇을 더할까?


```python
numbers = [1, 2, 3, 4]
print(sum(1 for x in numbers if x % 2 == 0))  # 2
print(sum(x for x in numbers if x % 2 == 0))  # 6
```

### 7. 리스트 [1, 2, 3]에서 [1:3]으로 자르면 어떤 값이 남을까?


```python
print([1, 2, 3][1:3])  # [2, 3]
```

### 8. 음수의 몫 -7 // 2는 얼마일까?


```python
print(-7 // 2)  # -4
```

### 9. isqrt(15)는 무엇을 반환하며 완전제곱수인지는 어떻게 확인할까?


```python
from math import isqrt
n = 15
r = isqrt(n)
print(r)  # 3
print(r * r == n)  # False
```

### 10. Counter에서 없는 키를 조회하면?


```python
from collections import Counter
counts = Counter('apple')
print(counts['z'])  # 0
```

### 11. 중복을 없애되 처음 나온 순서를 유지하려면?


```python
numbers = [3, 1, 3, 2, 1]
print(list(dict.fromkeys(numbers)))  # [3, 1, 2]
```
