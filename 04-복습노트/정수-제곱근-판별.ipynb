## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
from math import isqrt

def solution_model(n):
    x = isqrt(n)
    if x * x == n:
        return (x + 1) ** 2
    return -1

print(solution_model(121))  # 144
print(solution_model(3))    # -1
```
