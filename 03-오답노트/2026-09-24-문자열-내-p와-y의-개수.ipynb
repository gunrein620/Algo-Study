## 해설

### 풀이 메모

내 첫 풀이도 정답: p/P, y/Y를 각각 세고 두 개수를 비교했다.

lower로 소문자로 통일하면 대소문자를 따로 검사할 필요가 없다.

lower는 새 문자열을 반환하므로 s에 다시 저장한다.


## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(s):
    s = s.lower()
    return s.count('p') == s.count('y')

print(solution_model('pPoooyY'))  # 예상 출력: True
print(solution_model('Pyy'))  # 예상 출력: False
```

### 기존 예제와 보충 코드

이전에 기록한 경계 입력, 표현 예제, 비교 풀이입니다.


```python
def my_solution(s):
    cnt_p = 0
    cnt_y = 0
    for ch in s:
        if ch == 'p' or ch == 'P':
            cnt_p += 1
        if ch == 'y' or ch == 'Y':
            cnt_y += 1
    return cnt_p == cnt_y

print(solution_model("pPoooyY"))  # True

print(solution_model("Pyy"))     # False

print(solution_model("abc"))     # True: 둘 다 없으면 0 == 0
```
