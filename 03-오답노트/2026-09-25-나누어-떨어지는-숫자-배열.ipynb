## 해설

### 풀이 메모

내 실수: [i for i in arr if 조건 else -1]은 문법 오류.

뒤쪽 if는 걸러내기: [i for i in arr if 조건]

앞쪽 if/else는 값 바꾸기: [i if 조건 else -1 for i in arr]

이 문제는 원소마다 -1을 넣는 게 아니라 결과 전체가 비었을 때 [-1].

a or b: a가 거짓으로 평가되면 b, 아니면 a를 반환한다.

[]는 거짓, [0]은 비어 있지 않아서 참.


## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(arr, divisor):
    answer = sorted([i for i in arr if i % divisor == 0])
    return answer or [-1]

print(solution_model([5, 9, 7, 10], 5))  # 예상 출력: [5, 10]
print(solution_model([2, 36, 1, 3], 1))  # 예상 출력: [1, 2, 3, 36]
print(solution_model([3, 2, 6], 10))  # 예상 출력: [-1]
```
