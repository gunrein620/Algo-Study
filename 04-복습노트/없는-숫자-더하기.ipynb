## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(numbers):
    return 45 - sum(numbers)

print(solution_model([1, 2, 3, 4, 6, 7, 8, 0]))  # 14
print(solution_model([5, 8, 4, 0, 6, 7, 9]))     # 6
```

### 기존 풀이 비교 1

기존 노트에 기록된 다른 풀이입니다.


```python
def solution_compare_1(numbers):
    answer = 0
    for num in range(10):
        if num not in numbers:
            answer += num
    return answer

print(solution_compare_1([5, 8, 4, 0, 6, 7, 9]))  # 6
```
