# map — 원소마다 같은 함수를 적용하기

[전체 목차](../../README.md) · [도구 목차](../README.md) · [이 장 목차](README.md)

**이 장의 목표:** `list(map(int, text))`를 각 문자 → 정수 → 리스트 순서로 읽는다.
완성된 코드를 먼저 보고 중간 값으로 풀어봅니다. 각 코드 셀은 따로 실행해도 됩니다.

## 1. 개념과 모범 코드 — 하나씩 꺼내 같은 함수에 넣는다

`map(int, '1234')`는 문자열에서 문자를 하나씩 꺼내 각각 `int`에 넣습니다.
`list`로 감싸면 변환 결과가 리스트에 모입니다.

**이건 코테에서 자주 쓰니 챙겨라: `map(int, 대상)`은 대상의 원소마다 int를 적용한다.**


```python
text = '1234'
digits = list(map(int, text))
print(digits)  # 예상 출력: [1, 2, 3, 4]
print(type(digits[0]).__name__)  # 예상 출력: int
print(text)  # 예상 출력: 1234
```

| 꺼낸 원소(str) | 적용한 함수 | 나온 값(int) |
| --- | --- | --- |
| `'1'` | `int('1')` | 1 |
| `'2'` | `int('2')` | 2 |
| `'3'` | `int('3')` | 3 |
| `'4'` | `int('4')` | 4 |

문자열 전체를 정수 하나로 만드는 `int('1234')`와 다릅니다.
대상이 문자열이면 **한 글자씩**, 대상이 리스트이면 **리스트 원소 하나씩** 읽습니다.


```python
print(int('1234'))  # 예상 출력: 1234
print(list(map(int, '1234')))  # 예상 출력: [1, 2, 3, 4]
print(list(map(int, ['12', '34'])))  # 예상 출력: [12, 34]
```

마지막 줄에서 첫 원소는 `'1'`이 아니라 `'12'`입니다. 그래서 첫 변환 결과가 12입니다.

## 2. 두 인수의 역할 — 함수와 대상

기본 문법은 `map(함수, iterable)`입니다. 입력한 함수는 원소 하나를 인수로 받아야 합니다.

| 자리 | 이 예제의 값 | 역할 |
| --- | --- | --- |
| 첫 번째 | `int` | 나중에 원소마다 호출할 함수 자체 |
| 두 번째 | `'1234'` | 하나씩 꺼낼 대상 |
| 반환 | map 객체 | 변환 결과를 차례로 내주는 iterator |

`int`에는 괄호를 붙이지 않습니다. `int()`를 쓰면 지금 먼저 실행되어 0이 되고,
`map(0, 대상)` 꼴이 되어 결과를 꺼낼 때 `TypeError`가 납니다.
`map(int, 1234)`도 정수를 직접 순회할 수 없어서 `TypeError`입니다. 자릿수라면 `str(1234)`부터 필요합니다.

직접 만든 함수도 같은 규칙으로 전달할 수 있습니다.


```python
def double_model(x):
    return x * 2

print(list(map(double_model, [2, 4, 6])))  # 예상 출력: [4, 8, 12]
```

## 3. 일반 for문으로 풀어보면

처음 코드의 계산을 직접 쓰면 **꺼내기 → 바꾸기 → 추가하기**입니다.


```python
text = '1234'
digits = []
for ch in text:
    number = int(ch)
    digits.append(number)
print(digits)  # 예상 출력: [1, 2, 3, 4]
```

이때 `ch`는 str, `number`는 int, `digits`는 list입니다.
`map`은 이 반복을 짧게 표현하지만, 결과를 모으는 역할은 별도의 `list`가 합니다.

## 4. 결과가 리스트인지 합계인지에 따라 바깥 도구를 고른다

필요한 것이 합계라면 `sum`이 map의 결과를 바로 읽을 수 있습니다.


```python
n = 305
digits = list(map(int, str(n)))
total = sum(map(int, str(n)))
print(digits)  # 예상 출력: [3, 0, 5]
print(total)  # 예상 출력: 8
print(type(total).__name__)  # 예상 출력: int
```

| 필요한 결과 | 표현 | 역할 |
| --- | --- | --- |
| 자릿수 리스트 | `list(map(int, str(n)))` | 전부 저장 |
| 자릿수 합 | `sum(map(int, str(n)))` | 읽으면서 합산 |

`sum(list(map(...)))`도 가능하지만 합만 필요할 때 중간 리스트를 만들 이유는 없습니다.

### 모범 코드 — 자릿수 합으로 나누어떨어지는지

조건은 **양의 정수 x**입니다. 0은 자릿수 합도 0이라 나눗셈의 기준으로 쓸 수 없습니다.


```python
def harshad_model(x):
    digit_sum = sum(map(int, str(x)))
    return x % digit_sum == 0

print(harshad_model(18))  # 예상 출력: True
print(harshad_model(19))  # 예상 출력: False
```

| 단계 | x=18일 때 값 | 타입 |
| --- | --- | --- |
| `str(x)` | `'18'` | str |
| map이 제공하는 원소 | `1`, `8` | 각각 int |
| `digit_sum` | `9` | int |
| `x % digit_sum` | `0` | int |
| `0 == 0` | `True` | bool |

x를 나누어 갱신하지 않으므로 원래 18이 마지막 나머지 계산까지 그대로 남습니다.
한 줄이 낯설 때의 같은 계산은 다음과 같습니다.


```python
x = 18
digit_sum = 0
for ch in str(x):
    digit_sum += int(ch)
answer = x % digit_sum == 0
print(digit_sum)  # 예상 출력: 9
print(x)  # 예상 출력: 18
print(answer)  # 예상 출력: True
```

## 5. 반대 방향 — 숫자들을 문자열로 바꿔 붙이기

`join`은 문자열 원소만 받습니다. 정수 리스트라면 `map(str, numbers)`로 원소를 먼저 바꿉니다.


```python
numbers = [3, 0, 5]
text = ''.join(map(str, numbers))
print(text)  # 예상 출력: 305
print(type(text).__name__)  # 예상 출력: str
print(numbers)  # 예상 출력: [3, 0, 5]
```

흐름은 `[3, 0, 5]` → `'3', '0', '5'` → `'305'`입니다.
`str(numbers)`는 `'[3, 0, 5]'`라는 문자열을 만들기 때문에 이 목적과 다릅니다.
최종 답이 정수여야 한다면 **붙인 뒤** `int(text)`를 적용합니다. [join 자세히](../02-문자열/join.md)

## 6. map 결과는 한 번 소비한다

map은 만들자마자 변환 결과 전체를 저장하지 않습니다. `list`, `sum`, `for`, `next`가 요청할 때 변환합니다.


```python
source = ['1', '2', '3']
converted = map(int, source)
print(type(converted).__name__)  # 예상 출력: map
print(list(converted))  # 예상 출력: [1, 2, 3]
print(list(converted))  # 예상 출력: []
print(source)  # 예상 출력: ['1', '2', '3']

# 여러 번 써야 하면 결과를 한 번 모아 둡니다.
saved = list(map(int, source))
print(sum(saved))  # 예상 출력: 6
print(sum(saved))  # 예상 출력: 6
```

원본 리스트가 비워진 것이 아니라 **같은 map 객체에서 꺼낼 결과가 끝난 것**입니다.
입력이 다른 iterator라면 그 iterator도 소비합니다. [소비를 눈으로 확인하기](iterable-iterator.ipynb)

## 7. 빈 입력·잘못된 값·도구 선택


```python
print(list(map(int, '')))  # 예상 출력: []
print(sum(map(int, [])))  # 예상 출력: 0
print(''.join(map(str, [])) == '')  # 예상 출력: True
```

`list(map(int, '12a'))`는 `'a'`를 정수로 바꿀 때 `ValueError`입니다.
map 객체를 만드는 시점이 아니라 **그 결과를 꺼내는 시점**에 변환 오류가 나타납니다.
map 자체는 입력 컬렉션을 수정하지 않습니다. 다만 전달한 함수가 객체를 수정하는 함수라면 그 함수의 동작은 별개입니다.

| 상황 | 고를 표현 |
| --- | --- |
| 이미 있는 함수를 모든 원소에 적용 | `list(map(int, texts))` |
| 간단한 계산식으로 새 값 만들기 | `[x * 2 for x in numbers]` |
| 조건에 맞는 값만 고르기 | `[x for x in numbers if x > 0]` |
| 분기나 중간 설명이 길어짐 | 일반 for문 |

map 자체는 필터가 아닙니다. 조건에 맞지 않는 값을 자동으로 버리지 않습니다.
여러 iterable을 `map(함수, a, b)`처럼 줄 수도 있습니다. 이 경우 함수는 원소 두 개를 받고,
짧은 입력이 끝나면 멈춥니다. 지금은 한 대상 변환을 먼저 익히고, 두 배열의 관계는 다음 장의 zip으로 읽어도 충분합니다.

## 8. 가져갈 것

1. `map(int, 대상)`의 int는 함수 자체다. `int()`가 아니다.
2. map의 결과를 **모으려면 list, 더하려면 sum, 글자로 붙이려면 join**을 고른다.
3. 같은 map 객체는 끝까지 읽고 나면 비어 있다.

n개 원소, 변환 함수 한 번의 비용 T일 때 전체 변환 시간은 O(nT)입니다.
map 자체의 추가 공간은 O(1), list로 결과를 모으면 O(n)입니다(원소 크기를 일정하게 볼 때).
자릿수 합 예제에서는 `str(x)`가 자릿수 d만큼 문자열 공간 O(d)을 사용합니다.
