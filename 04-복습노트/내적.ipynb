## 해설

- 기록일: 2026-09-25
- 문제: 길이가 같은 두 정수 배열의 같은 위치끼리 곱한 값을 모두 더하기.
- 학습 증거: 음양 더하기에서 접한 zip을 이번 문제에 직접 선택해서 올바른 반복문 풀이를 작성함.
- 플랫폼 제출 통과와 이후 독립 재풀이는 미확인.

### 답을 보기 전에

1. zip(a, b)는 모든 조합을 만드는가, 같은 위치끼리 묶는가?
2. `for a, b in temp`가 동작해도 변수명을 바꾸는 편이 좋은 이유는?
3. 누적 반복문을 sum과 제너레이터로 어떻게 표현할까?

**같은 위치끼리 묶기 → 각각 곱하기 → 전부 더하기**입니다.
제너레이터가 곱을 하나씩 내놓고 sum이 합산하므로 중간 리스트는 필요 없습니다.

temp를 만들 때 이미 두 배열로 zip 객체를 생성했으므로, 반복 변수에 a와 b를 다시 사용해도 이 코드는 동작합니다.
다만 반복 중 함수 안의 a, b는 배열 대신 정수를 가리킵니다. 배열 원소를 수정하는 것은 아니지만 이후 a나 b를 배열로 사용하려 하면 혼란이 생깁니다.

| zip이 꺼내는 튜플 | x * y | 누적 합 |
| --- | --- | --- |
| (1, -3) | -3 | -3 |
| (2, -1) | -2 | -5 |
| (3, 0) | 0 | -5 |
| (4, 2) | 8 | 3 |

`for x, y`는 튜플의 두 값을 각각 x와 y에 나누어 담습니다. 여기서 x와 y는 인덱스가 아니라 정수 원소입니다.

### 음양 더하기와 연결

| 문제 | 같은 위치끼리 묶은 뒤 할 일 |
| --- | --- |
| 음양 더하기 | 숫자와 부호를 묶고 양수 또는 음수 선택 |
| 내적 | 숫자 두 개를 묶고 곱하기 |

두 문제 모두 변환한 값들을 마지막에 합산합니다. 길이가 같다고 보장하므로 zip이 짧은 쪽에서 끝나는 동작도 문제가 되지 않습니다.
배열 길이가 n일 때 시간 O(n), 추가 공간 O(1)입니다. 입력 배열은 변경하지 않습니다.

### 챙길 것

1. 두 배열의 같은 위치를 함께 처리할 때 zip을 떠올린다.
2. 계산한 값들의 합은 `sum(계산식 for ...)`으로 표현할 수 있다.
3. 입력 배열과 원소의 변수명을 구분한다: a/b와 x/y.

## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(a, b):
    return sum(x * y for x, y in zip(a, b))

print(solution_model([1, 2, 3, 4], [-3, -1, 0, 2]))  # 3
print(solution_model([-1, 0, 1], [1, 0, -1]))        # -2
```

### 기존 풀이 비교 1

기존 노트에 기록된 다른 풀이입니다.


```python
def solution_compare_1(a, b):
    answer = 0
    temp = zip(a, b)
    for a, b in temp:
        answer += a * b
    return answer

print(solution_compare_1([1, 2, 3, 4], [-3, -1, 0, 2]))  # 3
```

### 기존 풀이 비교 2

기존 노트에 기록된 다른 풀이입니다.


```python
def solution_compare_2(a, b):
    answer = 0
    for x, y in zip(a, b):
        answer += x * y
    return answer

print(solution_compare_2([1, 2, 3, 4], [-3, -1, 0, 2]))  # 3
print(solution_compare_2([0], [-1000]))                  # 0
```
