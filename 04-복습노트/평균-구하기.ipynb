## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(arr):
    return sum(arr) / len(arr)

print(solution_model([1, 2, 3, 4]))  # 2.5
print(solution_model([5, 5]))        # 5.0
```
