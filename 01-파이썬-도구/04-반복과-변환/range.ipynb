# range — 필요한 정수만 순서대로 만들기

[전체 목차](../../README.md) · [도구 목차](../README.md) · [이 장 목차](README.md)

**이 장의 목표:** `range`가 내놓는 숫자와, 그 숫자로 접근한 리스트의 값을 구분한다.
개념 → 모범 코드 → 값 추적 순서로 읽으세요. 각 코드 셀은 따로 실행해도 됩니다.
출력은 비워 두었으며 코드 옆의 `# 예상 출력:`과 실행 결과를 비교하면 됩니다.

## 1. 개념과 모범 코드 — 끝 숫자 바로 전까지

`range(1, 5)`는 **1부터 시작해 5에 도달하기 전까지** 정수를 제공합니다.
따라서 나오는 값은 `1, 2, 3, 4`입니다. `range`는 리스트가 아니라 정수 범위를 표현하는 객체입니다.

**이건 코테에서 자주 쓰니 챙겨라: `range(시작, 끝)`에서 끝은 포함하지 않는다.**


```python
numbers = list(range(1, 5))
print(numbers)  # 예상 출력: [1, 2, 3, 4]
print(sum(range(1, 5)))  # 예상 출력: 10
print(type(range(1, 5)).__name__)  # 예상 출력: range
```

`list(...)`는 안에서 나온 값을 리스트에 모으고, `sum(...)`은 그 값을 더합니다.
`type(값).__name__`은 타입 이름을 확인하는 학습용 표현입니다.

| 순서 | 나온 값 | 합계 |
| --- | --- | --- |
| 1 | 1 | 1 |
| 2 | 2 | 3 |
| 3 | 3 | 6 |
| 4 | 4 | 10 |
| 종료 | 다음 값 5는 끝 경계이므로 제외 | 10 |

합계를 직접 누적하면 같은 흐름을 볼 수 있습니다.


```python
total = 0
for number in range(1, 5):
    total += number
print(total)  # 예상 출력: 10
```

## 2. 인수는 시작·끝·간격이다

| 문법 | 해석 | `list(...)`로 확인한 값 |
| --- | --- | --- |
| `range(4)` | 0부터, 4 미만, 1씩 | `[0, 1, 2, 3]` |
| `range(2, 5)` | 2부터, 5 미만, 1씩 | `[2, 3, 4]` |
| `range(0, 7, 3)` | 0부터, 7 미만, 3씩 | `[0, 3, 6]` |
| `range(4, -1, -1)` | 4부터, -1 초과, 1씩 감소 | `[4, 3, 2, 1, 0]` |

인수를 하나만 주면 그것은 **끝**입니다. 시작의 기본값은 0, 간격의 기본값은 1입니다.
간격이 양수면 `현재 값 < 끝`, 음수면 `현재 값 > 끝`일 동안 진행합니다.


```python
print(list(range(4)))  # 예상 출력: [0, 1, 2, 3]
print(list(range(2, 5)))  # 예상 출력: [2, 3, 4]
print(list(range(0, 7, 3)))  # 예상 출력: [0, 3, 6]
print(list(range(4, -1, -1)))  # 예상 출력: [4, 3, 2, 1, 0]
```

`range(4, -1, -1)`의 두 `-1`은 역할이 다릅니다.
두 번째 인수는 **도달하면 멈출 경계**, 세 번째 인수는 **한 번 이동할 양**입니다.
0도 꺼내려면 경계를 0보다 작게 둬야 합니다.

## 3. 인덱스와 값 — range가 알아서 리스트를 읽지는 않는다

`arr = [10, 20, 30]`에서 위치는 `0, 1, 2`, 저장된 값은 `10, 20, 30`입니다.
`range(len(arr))`가 만드는 것은 위치로 쓸 수 있는 숫자입니다. 값을 읽는 동작은 `arr[i]`가 합니다.


```python
arr = [10, 20, 30]
indices = list(range(len(arr)))
values = []
for i in range(len(arr)):
    values.append(arr[i])
print(indices)  # 예상 출력: [0, 1, 2]
print(values)  # 예상 출력: [10, 20, 30]
```

| 반복 | `i` — 위치(int) | `arr[i]` — 값(int) |
| --- | --- | --- |
| 1 | 0 | 10 |
| 2 | 1 | 20 |
| 3 | 2 | 30 |

값만 필요하다면 `range`를 끼울 필요가 없습니다.


```python
arr = [10, 20, 30]
total = 0
for value in arr:
    total += value
print(total)  # 예상 출력: 60
```

| 필요한 것 | 고를 표현 |
| --- | --- |
| 값만 | `for value in arr` |
| 위치만 | `for i in range(len(arr))` |
| 위치와 값 모두 | `for i, value in enumerate(arr)` |

`range(arr)`는 리스트를 숫자 경계로 줄 수 없어서 `TypeError`입니다.
`for i in len(arr)`도 실패합니다. `len(arr)`는 정수 3이고, 정수 자체는 순회할 수 없습니다.

## 4. 모범 코드 — 길이가 정해진 부분문자열 순회

문자열 `t`에서 길이 `length`인 조각을 왼쪽부터 모두 모읍니다.
조건은 `1 ≤ length ≤ len(t)`입니다. **마지막으로 시작할 수 있는 위치**를 먼저 구하면 됩니다.


```python
def substrings_model(t, length):
    pieces = []
    for i in range(len(t) - length + 1):
        pieces.append(t[i:i + length])
    return pieces

print(substrings_model('31415', 3))  # 예상 출력: ['314', '141', '415']
print(substrings_model('abc', 3))  # 예상 출력: ['abc']
```

문자열 길이 5에서 길이 3을 잘라야 하므로 마지막 시작 위치는 `5 - 3 = 2`입니다.
시작 위치 2까지 포함하려면 `range(3)`이어야 해서 **`+ 1`**이 붙습니다.

| `i` | 자르는 표현 | 결과(str) |
| --- | --- | --- |
| 0 | `t[0:3]` | `'314'` |
| 1 | `t[1:4]` | `'141'` |
| 2 | `t[2:5]` | `'415'` |

`range`의 끝과 슬라이싱의 끝은 둘 다 미포함입니다. `i`는 조각이 아니라 **조각의 시작 위치**입니다.
슬라이싱 자체는 [인덱싱·슬라이싱](../06-문법/slicing.md)에서 이어서 봅니다.

## 5. 빈 범위와 잘못된 간격

시작부터 경계 조건에 맞지 않으면 빈 범위입니다. 자동으로 방향을 바꾸지 않습니다.


```python
print(list(range(0)))  # 예상 출력: []
print(list(range(4, 0)))  # 예상 출력: []
print(list(range(1, 5, -1)))  # 예상 출력: []
print(list(range(4, 0, -1)))  # 예상 출력: [4, 3, 2, 1]

total = 0
for number in range(0):
    total += number
print(total)  # 예상 출력: 0
```

| 입력 | 동작 |
| --- | --- |
| `range(1, 5, 0)` | 이동할 수 없으므로 `ValueError` |
| `range(3.5)` | 실수 경계는 받지 않아 `TypeError` |
| `range()` | 끝 인수가 없어 `TypeError` |

입력은 보통 정수입니다. 더 정확히는 정수 인덱스로 해석할 수 있는 객체를 받습니다.
반환은 다시 순회할 수 있는 `range`이며 원본 데이터를 바꾸는 연산이 아닙니다.

## 6. 가져갈 것

1. `range(n)`은 `0`부터 `n - 1`까지다. 결과가 n개의 정수다(단, n ≥ 0).
2. `i`와 `arr[i]`를 구분한다. 값만 필요하면 리스트를 직접 순회한다.
3. 마지막 위치를 **포함**하려면 끝 경계에 1을 더한다.

일반적인 코테의 고정 크기 정수 연산을 기준으로, 원소가 n개인 `range` 생성은 시간·공간 O(1),
전체 순회는 O(n)입니다. `list(range(...))`로 모으면 시간·공간 O(n)이 듭니다.
부분문자열 예제는 문자열 길이 N, 조각 길이 L일 때 각 조각 복사 때문에 시간·결과 공간 O((N-L+1)L)입니다.
