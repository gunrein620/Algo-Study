# 타입 변환 — 문자·숫자·리스트를 구분해서 쓰기

[전체 목차](../../README.md) · [도구 목차](../README.md) · [이 장 목차](README.md)

**이 장의 목표:** 변환 전후의 값과 타입을 따라가며 `int`, `str`, `list`, `tuple`을 고른다.
각 코드 셀은 독립 실행할 수 있습니다. 출력 모양만 보지 말고 타입도 함께 확인하세요.

## 1. 개념과 모범 코드 — 숫자처럼 생긴 문자도 문자열이다

`'12'`는 문자열(str), `12`는 정수(int)입니다. 계산하려면 `int`, 글자로 다루려면 `str`을 씁니다.

**이건 코테에서 자주 쓰니 챙겨라: 계산할 때는 `int`, 자릿수를 글자로 읽을 때는 `str`이다.**


```python
text = '12'
number = int(text)
print(number + 3)  # 예상 출력: 15
print(text + '3')  # 예상 출력: 123
print(type(text).__name__)  # 예상 출력: str
print(type(number).__name__)  # 예상 출력: int
print('1' == 1)  # 예상 출력: False
```

같은 `+`라도 정수끼리는 덧셈, 문자열끼리는 이어 붙이기입니다.
`'12' + 3`처럼 둘을 섞으면 Python이 마음대로 타입을 맞춰 주지 않아 `TypeError`가 납니다.

| 표현 | 결과 | 타입 | 의미 |
| --- | --- | --- | --- |
| `int('12')` | `12` | int | 문자열 전체를 수로 해석 |
| `str(12)` | `'12'` | str | 수를 글자로 표현 |
| `list('12')` | `['1', '2']` | list, 원소는 str | 한 글자씩 모으기 |
| `tuple('12')` | `('1', '2')` | tuple, 원소는 str | 한 글자씩 튜플에 모으기 |

## 2. int — 문자열 전체를 정수로 읽는다

`int(x)`는 정수로 해석 가능한 문자열이나 수를 받아 int를 반환합니다.
`int(문자열, 진법)`은 해당 진법으로 적힌 문자열을 정수로 읽습니다.


```python
print(int('123'))  # 예상 출력: 123
print(int('-123'))  # 예상 출력: -123
print(int('007'))  # 예상 출력: 7
print(int('  12  '))  # 예상 출력: 12
print(int(-3.9))  # 예상 출력: -3
print(int('101', 2))  # 예상 출력: 5
print(int('120', 3))  # 예상 출력: 15
```

`int('120', 3)`은 `1 × 3² + 2 × 3¹ + 0 × 3⁰ = 15`입니다.
진법을 생략하면 문자열은 기본적으로 10진수로 읽습니다. 이 장에서는 2~36진법 중 필요한 진법만 사용합니다.
자세한 적용은 [3진법 뒤집기 기록](../../03-오답노트/2026-09-26-3진법-뒤집기.ipynb)으로 연결합니다.

실수를 넣으면 반올림하지 않고 **0 방향으로 소수 부분을 버립니다**. 그래서 `int(-3.9)`는 -3입니다.
`int('3.9')`는 다릅니다. 정수 표현이 아닌 문자열이라 `ValueError`입니다.

| 입력 | 결과 또는 예외 |
| --- | --- |
| `int()` | `0` |
| `int('')`, `int('abc')`, `int('3.9')` | `ValueError` |
| `int('102', 2)` | 2진법에 숫자 2가 없어서 `ValueError` |
| `int(None)`, `int([1, 2])` | 받을 수 없는 타입이어서 `TypeError` |

## 3. str — 숫자를 글자로 바꾸면 자릿수를 읽을 수 있다

`str(x)`는 객체의 문자열 표현을 반환합니다. 원본 객체의 타입을 직접 바꾸지는 않습니다.


```python
n = 305
text = str(n)
chars = list(text)
print(text)  # 예상 출력: 305
print(type(text).__name__)  # 예상 출력: str
print(chars)  # 예상 출력: ['3', '0', '5']
print(n)  # 예상 출력: 305
print(type(n).__name__)  # 예상 출력: int
```

`print(305)`와 `print('305')`는 화면에 똑같이 `305`를 표시합니다.
**따옴표가 안 보인다고 정수라고 판단하면 안 됩니다.** `type(...)`으로 확인하세요.

정수에는 글자의 첫 위치라는 개념이 없으므로 `n[0]`은 `TypeError`입니다.
`str(n)[0]`은 먼저 문자열로 바꾼 다음 그 문자열의 첫 글자를 읽습니다.


```python
n = 305
first = str(n)[0]

# 같은 표현을 중간 변수로 풀기
text = str(n)
first_stepwise = text[0]
print(first)  # 예상 출력: 3
print(type(first).__name__)  # 예상 출력: str
print(first_stepwise)  # 예상 출력: 3
print(str([1, 2]))  # 예상 출력: [1, 2]
print(type(str([1, 2])).__name__)  # 예상 출력: str
```

`str([1, 2])`는 대괄호와 쉼표까지 포함한 문자열 `'[1, 2]'`입니다.
원소를 붙인 `'12'`가 필요하면 [map과 join 연결](map.ipynb)을 사용합니다.

## 4. list와 tuple — 원소의 타입까지 바꾸지는 않는다

`list(iterable)`은 원소를 리스트에 모으고, `tuple(iterable)`은 튜플에 모읍니다.
튜플은 원소의 위치를 교체하거나 원소를 추가·삭제할 수 없는 순서열입니다.


```python
chars = list('123')
numbers = [1, 2, 3]
fixed = tuple(numbers)
print(chars)  # 예상 출력: ['1', '2', '3']
print(type(chars[0]).__name__)  # 예상 출력: str
print(fixed)  # 예상 출력: (1, 2, 3)
print(list(fixed))  # 예상 출력: [1, 2, 3]
print(numbers)  # 예상 출력: [1, 2, 3]
```

`list('123')`은 글자를 모았으므로 원소가 str입니다. 숫자 원소가 필요하면 **각 원소에 int를 적용**해야 합니다.
이 작업을 짧게 표현하는 도구가 다음 장의 `map`입니다.


```python
digits = list(map(int, '123'))
print(digits)  # 예상 출력: [1, 2, 3]
print(type(digits[0]).__name__)  # 예상 출력: int

# 일반 for문으로 같은 변환
digits_stepwise = []
for ch in '123':
    digits_stepwise.append(int(ch))
print(digits_stepwise)  # 예상 출력: [1, 2, 3]
```

| 비교 | 역할 |
| --- | --- |
| `int('123')` | 문자열 전체 → 정수 하나 123 |
| `list('123')` | 문자열의 각 문자 → 문자 리스트 |
| `list(map(int, '123'))` | 각 문자에 int 적용 → 정수 리스트 |

## 5. 모범 코드 — 자연수의 자릿수를 뒤집어 정수 리스트로

조건은 `n ≥ 0`인 정수입니다. 음수는 `'-'`까지 문자로 포함되므로 같은 코드에 그대로 넣지 않습니다.


```python
def reversed_digits_model(n):
    return list(map(int, str(n)[::-1]))

print(reversed_digits_model(305))  # 예상 출력: [5, 0, 3]
print(reversed_digits_model(0))  # 예상 출력: [0]
```

한 줄을 안쪽부터 풀면 **문자열로 → 뒤집기 → 문자마다 정수로 → 리스트로**입니다.


```python
n = 305
text = str(n)
backward = text[::-1]
digits = []
for ch in backward:
    digits.append(int(ch))
print(text)  # 예상 출력: 305
print(backward)  # 예상 출력: 503
print(digits)  # 예상 출력: [5, 0, 3]
```

| 단계 | 값 | 타입 |
| --- | --- | --- |
| 입력 | `305` | int |
| `str(n)` | `'305'` | str |
| `text[::-1]` | `'503'` | str |
| 글자 하나 읽기 | `'5'`, `'0'`, `'3'` | 각각 str |
| 글자마다 `int(ch)` | `5`, `0`, `3` | 각각 int |
| 리스트에 모으기 | `[5, 0, 3]` | list |

## 6. 경계 조건과 원본 변경 여부

인수를 아예 생략하는 것과 빈 문자열을 정수로 바꾸는 것은 다릅니다.


```python
print(int())  # 예상 출력: 0
print(str() == '')  # 예상 출력: True
print(list())  # 예상 출력: []
print(tuple())  # 예상 출력: ()
print(list(''))  # 예상 출력: []
print(tuple(''))  # 예상 출력: ()
```

`list(123)`과 `tuple(123)`은 정수 하나를 순회할 수 없어 `TypeError`입니다.
`tuple([1])`의 결과는 `(1,)`입니다. 원소 하나인 튜플의 쉼표를 눈여겨보세요.

변환 도구 자체는 원본 리스트·문자열을 수정하지 않습니다. 다만 iterator를 list/tuple에 넣으면 남은 값을 소비합니다.
**보충:** `list(기존 리스트)`는 바깥 리스트만 새로 만드는 얕은 복사입니다.


```python
original = [[1], [2]]
copied = list(original)
copied.append([3])
print(original)  # 예상 출력: [[1], [2]]
print(copied)  # 예상 출력: [[1], [2], [3]]
copied[0].append(9)
print(original)  # 예상 출력: [[1, 9], [2]]
print(copied)  # 예상 출력: [[1, 9], [2], [3]]
```

바깥 리스트는 다르지만 안쪽 `[1]`은 함께 참조합니다. 원본이 변한 시점은 `list(original)`이 아니라
공유한 안쪽 리스트에 `append(9)`를 한 시점입니다. 튜플도 내부의 리스트까지 불변으로 만들지는 않습니다.

## 7. 가져갈 것

1. `'1'`과 `1`은 타입이 다르다. 출력 모양만으로 구분하지 않는다.
2. `list`는 원소를 모을 뿐, 문자 원소를 정수로 바꾸지는 않는다.
3. 한 줄이 낯설면 `text`, `backward`, `digits`처럼 중간 값을 이름 붙여 따라간다.

n개 원소를 list/tuple로 새로 모으면 보통 시간·공간 O(n)입니다. 이미 tuple인 값을 tuple에 넘기면 재사용할 수 있습니다.
숫자와 문자열 사이의 변환 비용은 자릿수와 진법에 따라 달라지므로 항상 O(1)이라고 두지 않습니다.
위 자릿수 모범 코드에서 문자열 변환 이후 처리는 자릿수 d에 대해 시간·결과 공간 O(d)입니다.
