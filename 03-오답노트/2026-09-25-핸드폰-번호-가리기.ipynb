## 해설

### 풀이 메모

내 반복문 풀이도 정답: reversed로 뒤에서 읽고 4개 보존, 다시 뒤집었다.

range는 정수 위치를 주고, reversed(문자열)는 문자 자체를 준다.

슬라이싱으로 마지막 4글자를 바로 가져올 수 있다.

"321432"의 인덱스:  0  1  2  3  4  5

음수 인덱스:       -6 -5 -4 -3 -2 -1

[-4:]는 뒤에서 네 번째부터 끝까지. 음수 위치가 역순을 뜻하진 않는다.

[::-1]의 -1은 간격이므로 역순이다.


## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(phone_number):
    return '*' * (len(phone_number) - 4) + phone_number[-4:]

print(solution_model('01033334444'))  # 예상 출력: *******4444
print(solution_model('027778888'))  # 예상 출력: *****8888
```

### 기존 예제와 보충 코드

이전에 기록한 경계 입력, 표현 예제, 비교 풀이입니다.


```python
s = "321432"

print(s[-4])    # 1: 문자 하나

print(s[-4:])   # 1432: 마지막 4글자

print(s[:-4])   # 32: 마지막 4글자 앞까지

print(s[::-1])  # 234123: 전체 역순

print(solution_model(s))       # **1432

print(solution_model("1234"))  # 1234
```
