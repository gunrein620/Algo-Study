## 해설

### 풀이 메모

내 실수: sorted(..., reverse=True)는 큰 순서 정렬이지 순서 뒤집기가 아니다.

str(n)도 문자열이므로 바로 [::-1]을 붙일 수 있다.

12345 → "12345" → "54321" → [5, 4, 3, 2, 1]


## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(n):
    return list(map(int, str(n)[::-1]))

print(solution_model(12345))  # 예상 출력: [5, 4, 3, 2, 1]
```

### 기존 예제와 보충 코드

이전에 기록한 경계 입력, 표현 예제, 비교 풀이입니다.


```python
print(solution_model(12345))  # [5, 4, 3, 2, 1]

print(solution_model(120))    # [0, 2, 1]
```
