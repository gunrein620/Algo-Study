## 해설

### 풀이 메모

막힌 점: 문자 코드로 이동하는 아이디어는 맞았지만, 끝에서 순환하는 법을 몰랐다.

핵심: (현재 위치 + 이동량) % 전체 개수. 위치는 0부터 시작한다.

chr는 자동으로 순환하지 않는다: chr(ord('z') + 1)은 '{'.

'z'를 2칸 이동: 122 - 97 = 25 → (25 + 2) % 26 = 1 → chr(97 + 1) = 'b'


## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(s, n):
    answer = []
    for ch in s:
        if ch == ' ':
            answer.append(ch)
            continue

        base = ord('A') if ch.isupper() else ord('a')
        idx = ord(ch) - base       # 문자 코드를 알파벳 위치 0~25로 바꾼다.
        moved = (idx + n) % 26     # 26은 0으로, 27은 1로 돌아온다.
        answer.append(chr(base + moved))  # 위치를 문자 코드로 되돌린 뒤 문자로.

    return ''.join(answer)

print(solution_model('AB', 1))  # 예상 출력: BC
print(solution_model('z', 1))  # 예상 출력: a
print(solution_model('a B z', 4))  # 예상 출력: e F d
```
