## 해설

### 풀이 메모

내 실수: 중첩 for로 모든 조합을 만들었다. 같은 위치끼리는 zip.

True는 양수, False는 음수. append[-num]이 아니라 append(-num).

짧게: sum(num if sign else -num for num, sign in zip(absolutes, signs))


## 모범 코드

모범 함수는 `solution_model`로 구분해 내가 작성한 `solution`을 덮어쓰지 않습니다. 제출할 때는 함수 이름을 `solution`으로 바꾸세요.


```python
def solution_model(absolutes, signs):
    answer = []
    for num, sign in zip(absolutes, signs):
        if sign:
            answer.append(num)
        else:
            answer.append(-num)
    return sum(answer)

print(solution_model([4, 7, 12], [True, False, True]))  # 예상 출력: 9
print(solution_model([1, 2, 3], [False, False, True]))  # 예상 출력: 0
```
