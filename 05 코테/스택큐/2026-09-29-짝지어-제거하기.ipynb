## 해설

### 직접 푼 접근

현재 문자와 스택의 마지막 문자가 같으면 `pop()`, 다르면 `append()`하는 풀이를 직접 작성했고 로직은 정답이었다.

처음에는 입력 문자열을 `list(s)`로 바꿨지만, 실제로 꺼내는 대상은 입력이 아니라 `stack`이다. 문자열 `s`는 한 글자씩 읽기만 할 수 있으므로 그대로 순회하면 된다.

### 핵심 아이디어

스택에는 지금까지 읽었지만 아직 짝을 만나지 못한 문자만 남긴다.

1. 스택이 비었거나 마지막 문자와 현재 문자가 다르면 현재 문자를 넣는다.
2. 마지막 문자와 현재 문자가 같으면 마지막 문자를 꺼내 둘을 제거한다.
3. 끝까지 처리한 뒤 스택이 비었으면 모두 제거된 것이다.

### 값 추적: `baabaa`

| 현재 문자 | 동작 | 처리 후 stack |
| --- | --- | --- |
| `b` | 추가 | `['b']` |
| `a` | 추가 | `['b', 'a']` |
| `a` | 같은 문자 제거 | `['b']` |
| `b` | 같은 문자 제거 | `[]` |
| `a` | 추가 | `['a']` |
| `a` | 같은 문자 제거 | `[]` |

### Python 도구

- 문자열은 불변이라 `s.pop()`을 사용할 수 없다.
- 리스트의 `pop()`은 마지막 값을 제거하고 그 값을 반환한다. 빈 리스트에서는 `IndexError`가 발생한다.
- `stack and stack[-1] == char`는 스택이 비어 있지 않을 때만 마지막 값을 확인한다.
- `return 0 if stack else 1`은 값이 남으면 실패, 비면 성공이라는 뜻이다.

시간 복잡도는 문자열 길이를 `n`이라 할 때 O(n), 추가 공간은 최악 O(n)이다.


## 모범 코드

직접 작성한 풀이와 같은 스택 로직이다. 입력 문자열은 그대로 순회하고, 별도 리스트만 스택으로 사용한다.


```python
def solution_model(s):
    stack = []

    for char in s:
        if stack and stack[-1] == char:
            stack.pop()
        else:
            stack.append(char)

    return 0 if stack else 1


print(solution_model('baabaa'))  # 1
print(solution_model('cdcd'))    # 0
```

### 가져갈 것

1. 입력 문자열은 읽기만 하고, 추가와 제거는 별도 스택에서 한다.
2. `stack and ...`로 빈 리스트 접근을 먼저 막는다.
3. 연속 제거 문제에서는 스택의 마지막 값과 현재 값을 비교한다.
