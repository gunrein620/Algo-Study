# enumerate — 인덱스와 값을 함께 읽기

[전체 목차](../../README.md) · [도구 목차](../README.md) · [이 장 목차](README.md)

**이 장의 목표:** `for i, value in enumerate(arr)`에서 두 변수에 무엇이 들어오는지 바로 읽는다.
개념과 완성된 코드를 먼저 읽고 값 추적 표로 확인하세요. 각 코드 셀은 독립 실행할 수 있습니다.

## 1. 개념과 모범 코드 — 값에 번호를 붙인다

`enumerate`는 원소를 읽으면서 **번호와 값 한 쌍**을 내줍니다. 기본 번호는 0부터입니다.
리스트에 같은 값이 여러 번 있어도 각각의 위치를 구분할 수 있습니다.

**이건 코테에서 자주 쓰니 챙겨라: 위치와 값이 둘 다 필요하면 `enumerate`다.**


```python
letters = ['a', 'b', 'a']
pairs = list(enumerate(letters))
print(pairs)  # 예상 출력: [(0, 'a'), (1, 'b'), (2, 'a')]

positions = []
for i, letter in enumerate(letters):
    if letter == 'a':
        positions.append(i)
print(positions)  # 예상 출력: [0, 2]
```

`enumerate(letters)`가 한 번에 내주는 것은 `(0, 'a')` 같은 **튜플(tuple)**입니다.
`for i, letter`는 그 튜플을 두 변수로 나누어 받습니다. 이를 **언패킹**이라고 합니다.

| 반복 | 받은 튜플 | `i` — 번호(int) | `letter` — 값(str) | 저장된 위치 |
| --- | --- | --- | --- | --- |
| 1 | `(0, 'a')` | 0 | `'a'` | `[0]` |
| 2 | `(1, 'b')` | 1 | `'b'` | `[0]` |
| 3 | `(2, 'a')` | 2 | `'a'` | `[0, 2]` |

`positions.append(letter)`로 바꾸면 위치가 아니라 문자 `'a'`가 저장됩니다.
문제가 **위치를 달라는지, 값을 달라는지** 먼저 확인하세요.

## 2. 낯선 for문을 두 단계로 풀기

쉼표로 받는 문법이 낯설면 먼저 한 쌍을 받은 뒤 나눠 보면 됩니다.


```python
pair = (2, 'a')
i, letter = pair
print(i)  # 예상 출력: 2
print(letter)  # 예상 출력: a
print(type(pair).__name__)  # 예상 출력: tuple
print(type(letter).__name__)  # 예상 출력: str
```

`enumerate` 없이 같은 위치 찾기를 구현하면 다음과 같습니다. 둘 중 외워야 할 코드는 위의 정석 풀이입니다.


```python
letters = ['a', 'b', 'a']
positions = []
for i in range(len(letters)):
    letter = letters[i]
    if letter == 'a':
        positions.append(i)
print(positions)  # 예상 출력: [0, 2]
```

두 코드 모두 i가 0, 1, 2일 때 letters의 값을 읽습니다.
`enumerate`를 쓰면 `letter = letters[i]`를 따로 적지 않아도 됩니다.

## 3. start는 번호의 시작이지, 읽기 시작할 위치가 아니다

문법은 `enumerate(iterable, start=0)`입니다. iterable은 리스트·문자열처럼 순회할 대상입니다.
start를 1로 바꾸면 **첫 번째 원소부터 읽으면서** 1, 2, 3이라는 번호를 붙입니다.


```python
names = ['민수', '지수', '현우']
print(list(enumerate(names)))  # 예상 출력: [(0, '민수'), (1, '지수'), (2, '현우')]
print(list(enumerate(names, start=1)))  # 예상 출력: [(1, '민수'), (2, '지수'), (3, '현우')]
print(names)  # 예상 출력: ['민수', '지수', '현우']
```

| 표현 | 첫 쌍 | 용도 |
| --- | --- | --- |
| `enumerate(names)` | `(0, '민수')` | 실제 0 기반 인덱스가 필요할 때 |
| `enumerate(names, start=1)` | `(1, '민수')` | 사람에게 보여줄 1번, 2번 번호 |
| `enumerate(names[1:], start=1)` | `(1, '지수')` | 슬라이싱으로 첫 원소를 제외하고 원래 위치에 맞게 번호 붙이기 |

`start=1`에서 받은 번호를 그대로 `names[번호]`에 쓰면 위치가 어긋납니다.
마지막 번호 3으로 `names[3]`을 읽으려 하면 `IndexError`입니다. **번호가 곧 원본 인덱스인 것은 start=0으로 원본을 그대로 읽을 때입니다.**

## 4. 모범 코드 — 문자열 위치에 따라 처리하기

문자열의 0, 2, 4, … 위치를 대문자로, 1, 3, 5, … 위치를 소문자로 만듭니다.
이 예제는 **하나의 문자열 전체 위치**가 기준입니다. 단어마다 위치를 초기화하는 문제라면 단어별 처리가 추가로 필요합니다.


```python
def alternate_case_model(text):
    answer = []
    for i, ch in enumerate(text):
        if i % 2 == 0:
            answer.append(ch.upper())
        else:
            answer.append(ch.lower())
    return ''.join(answer)

print(alternate_case_model('hello'))  # 예상 출력: HeLlO
print(repr(alternate_case_model('')))  # 예상 출력: ''
```

| `i` | `ch` | `i % 2` | 추가하는 문자 | answer |
| --- | --- | --- | --- | --- |
| 0 | `'h'` | 0 | `'H'` | `['H']` |
| 1 | `'e'` | 1 | `'e'` | `['H', 'e']` |
| 2 | `'l'` | 0 | `'L'` | `['H', 'e', 'L']` |
| 3 | `'l'` | 1 | `'l'` | `['H', 'e', 'L', 'l']` |
| 4 | `'o'` | 0 | `'O'` | `['H', 'e', 'L', 'l', 'O']` |

변경한 문자를 리스트에 모은 뒤 `''.join(answer)`로 하나의 문자열로 만듭니다.
문자열은 직접 수정할 수 없으므로 별도 결과를 만듭니다.
`repr(...)`는 빈 문자열도 따옴표 `''`가 보이게 표시하는 학습용 표현입니다.

## 5. index와 혼동하지 않기

`arr.index(값)`은 **처음 등장한 위치 하나**를 찾습니다. 현재 순회 위치를 알려주는 함수가 아닙니다.


```python
letters = ['a', 'b', 'a']
print(letters.index('a'))  # 예상 출력: 0
print(list(enumerate(letters)))  # 예상 출력: [(0, 'a'), (1, 'b'), (2, 'a')]
```

세 번째 `'a'`를 처리하는 중이어도 `letters.index('a')`는 0입니다.
반복 중 현재 위치가 필요하면 `enumerate`, 특정 값의 첫 위치만 찾으려면 [index](../03-리스트/index-find.md)를 고릅니다.

## 6. 반환 타입·소비·빈 입력

반환은 리스트가 아닌 **enumerate 객체(iterator)**입니다. 순회할 때마다 `(번호, 원소)` 튜플이 나옵니다.
번호는 int이고 원소는 원래 타입을 유지합니다. 원본 리스트를 수정하지는 않지만, 입력이 iterator라면 진행시켜 소비합니다.


```python
letters = ['x', 'y']
items = enumerate(letters)
print(type(items).__name__)  # 예상 출력: enumerate
print(list(items))  # 예상 출력: [(0, 'x'), (1, 'y')]
print(list(items))  # 예상 출력: []
print(letters)  # 예상 출력: ['x', 'y']
print(list(enumerate([])))  # 예상 출력: []
```

`enumerate(3)`은 정수 하나를 순회할 수 없어서 `TypeError`입니다.
start도 정수로 해석할 수 있어야 하므로 `enumerate(['a'], start=1.5)`는 `TypeError`입니다.
소비의 의미는 [iterable과 iterator](iterable-iterator.ipynb)에서 자세히 봅니다.

## 7. 가져갈 것

1. `for i, value in enumerate(arr)`는 **번호와 값**을 받는다.
2. `start=1`은 번호만 바꾼다. 첫 원소를 건너뛰지 않는다.
3. 반복 중 `index(value)`로 위치를 다시 찾지 않는다.

입력 원소 수를 n, 원소 하나를 읽는 비용을 O(1)로 보면 enumerate 생성은 O(1), 전체 순회는 O(n)입니다.
enumerate 자체의 추가 공간은 O(1), 결과를 list로 모으면 O(n)입니다.
위 두 모범 코드는 결과 저장까지 포함해 시간·공간 O(n)입니다(문자 예제는 영문자 기준).
